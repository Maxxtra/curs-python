### Tiparul „listă nouă”

```python
def peste_medie(note):
    media = sum(note) / len(note)
    rezultat = []
    for nota in note:
        if nota > media:
            rezultat.append(nota)
    return rezultat

print(peste_medie([9, 7, 10, 8, 6]))
```

## 4. Cum organizăm o aplicație

O aplicație adevărată are de obicei **trei părți**:

| Partea | Ce conține | Exemplu (catalog) |
|---|---|---|
| 🗂️ **Datele** | liste și dicționare | `catalog = [{"nume": ..., "nota": ...}]` |
| ⚙️ **Logica** | funcții care fac calcule, **fără `input`/`print`** | `media(catalog)`, `cel_mai_bun(catalog)` |
| 💬 **Interfața** | meniul, `input` și `print` | `while` cu „1. Adaugă, 2. Afișează…” |

Reguli simple:
1. **O funcție = o singură treabă**, cu un nume care spune ce face.
2. **Testezi fiecare funcție separat**, înainte să o pui în meniu.
3. Când lucrați în echipă, **fiecare ia o parte**: așa nu vă călcați pe picioare.

Mai jos e o aplicație mică, construită exact așa.

```python
# 🗂️ DATELE
catalog = []

# ⚙️ LOGICA
def adauga_elev(catalog, nume, nota):
    catalog.append({"nume": nume, "nota": nota})

def media(catalog):
    if len(catalog) == 0:
        return 0
    suma = 0
    for elev in catalog:
        suma += elev["nota"]
    return suma / len(catalog)

def cel_mai_bun(catalog):
    cel_mai_bun = catalog[0]
    for elev in catalog:
        if elev["nota"] > cel_mai_bun["nota"]:
            cel_mai_bun = elev
    return cel_mai_bun

# Testăm logica, înainte de meniu
adauga_elev(catalog, "Ana", 9)
adauga_elev(catalog, "Mihai", 7)
print(media(catalog))            # trebuie 8.0
print(cel_mai_bun(catalog))      # trebuie Ana
```

```python
# 💬 INTERFAȚA
def afiseaza_meniu():
    print()
    print("1. Adaugă elev")
    print("2. Afișează catalogul")
    print("3. Media și cel mai bun elev")
    print("0. Ieșire")

while True:
    afiseaza_meniu()
    optiune = input("Alege o opțiune: ")
    if optiune == "1":
        nume = input("Nume: ")
        nota = int(input("Nota: "))
        adauga_elev(catalog, nume, nota)
    elif optiune == "2":
        for elev in catalog:
            print(f"- {elev['nume']}: {elev['nota']}")
    elif optiune == "3":
        if catalog:
            print(f"Media: {media(catalog):.2f}")
            print(f"Cel mai bun: {cel_mai_bun(catalog)['nume']}")
        else:
            print("Catalogul e gol.")
    elif optiune == "0":
        print("La revedere!")
        break
    else:
        print("Opțiune necunoscută.")
```

> 🧑‍🏫 **Note pentru profesor**
>
> Arată că **meniul nu știe cum se calculează media**, doar cheamă funcția. Asta e ideea care face posibil lucrul în echipă: unul scrie `media`, altul scrie meniul, se întâlnesc la numele funcției. După-amiază vor împărți proiectul exact în aceste trei roluri.

### G1. Al doilea maxim
`al_doilea_maxim(lista)` returnează al doilea cel mai mare număr **distinct** sau `None` dacă nu există. Fără `sorted`, într-o singură parcurgere.

**Teste:** `[5, 9, 9, 7]` → `7` · `[3, 3]` → `None` · `[-1, -5]` → `-5` · `[4]` → `None`

```python
# G1. Al doilea maxim
```

```python
# Soluție – G1
def al_doilea_maxim(lista):
    primul = None
    al_doilea = None
    for x in lista:
        if primul is None or x > primul:
            al_doilea = primul
            primul = x
        elif x != primul and (al_doilea is None or x > al_doilea):
            al_doilea = x
    return al_doilea

print(al_doilea_maxim([5, 9, 9, 7]), al_doilea_maxim([3, 3]), al_doilea_maxim([-1, -5]), al_doilea_maxim([4]))
```

### G2. Rotirea
`roteste(lista, k)` returnează o listă **nouă**, rotită spre dreapta cu `k` poziții. `k` poate fi mai mare decât lungimea.

**Teste:** `([1, 2, 3, 4, 5], 2)` → `[4, 5, 1, 2, 3]` · `([1, 2, 3, 4, 5], 7)` → `[4, 5, 1, 2, 3]` · `([1, 2, 3], 0)` → `[1, 2, 3]` · `([], 3)` → `[]`

```python
# G2. Rotirea
```

```python
# Soluție – G2
def roteste(lista, k):
    if len(lista) == 0:
        return []
    k = k % len(lista)
    if k == 0:
        return lista[:]
    return lista[-k:] + lista[:-k]

print(roteste([1, 2, 3, 4, 5], 2), roteste([1, 2, 3, 4, 5], 7), roteste([1, 2, 3], 0), roteste([], 3))
```

### G3. Interclasarea
`interclaseaza(a, b)` primește două liste **sortate** și returnează lista sortată cu toate elementele, **fără** `sorted`/`sort`, parcurgând fiecare listă o singură dată (doi indici).

**Teste:** `([1, 4, 9], [2, 3, 10, 11])` → `[1, 2, 3, 4, 9, 10, 11]` · `([], [1, 2])` → `[1, 2]` · `([5, 5], [5])` → `[5, 5, 5]`

```python
# G3. Interclasarea
```

```python
# Soluție – G3
def interclaseaza(a, b):
    i = j = 0
    rezultat = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            rezultat.append(a[i])
            i += 1
        else:
            rezultat.append(b[j])
            j += 1
    return rezultat + a[i:] + b[j:]

print(interclaseaza([1, 4, 9], [2, 3, 10, 11]), interclaseaza([], [1, 2]), interclaseaza([5, 5], [5]))
```

### G4. Anagrame
`sunt_anagrame(a, b)` returnează `True` dacă cele două texte au aceleași litere, de același număr de ori, **ignorând spațiile și literele mari**. Folosiți un dicționar de frecvențe.

**Teste:** `("Listen", "Silent")` → `True` · `("roma", "amor")` → `True` · `("a gentleman", "elegant man")` → `True` · `("abc", "abd")` → `False` · `("aab", "abb")` → `False`

```python
# G4. Anagrame
```

```python
# Soluție – G4
def frecvente(text):
    d = {}
    for c in text.lower():
        if c != " ":
            d[c] = d.get(c, 0) + 1
    return d

def sunt_anagrame(a, b):
    return frecvente(a) == frecvente(b)

print(sunt_anagrame("Listen", "Silent"), sunt_anagrame("roma", "amor"), sunt_anagrame("a gentleman", "elegant man"), sunt_anagrame("abc", "abd"), sunt_anagrame("aab", "abb"))
```

### G5. Compresia
`comprima(text)` înlocuiește fiecare grup de litere identice consecutive cu litera urmată de numărul de repetări. `decomprima(text)` face invers (numărul poate avea mai multe cifre).

**Teste:** `comprima("aaabccdddd")` → `"a3b1c2d4"` · `comprima("")` → `""` · `decomprima("a3b1c2d4")` → `"aaabccdddd"` · `decomprima("x12y1")` → 12 de `x` și un `y`

```python
# G5. Compresia
```

```python
# Soluție – G5
def comprima(text):
    if text == "":
        return ""
    rezultat = ""
    curent = text[0]
    nr = 1
    for c in text[1:]:
        if c == curent:
            nr += 1
        else:
            rezultat += curent + str(nr)
            curent = c
            nr = 1
    return rezultat + curent + str(nr)

def decomprima(text):
    rezultat = ""
    litera = ""
    numar = ""
    for c in text:
        if c.isdigit():
            numar += c
        else:
            if litera != "":
                rezultat += litera * int(numar)
            litera = c
            numar = ""
    if litera != "":
        rezultat += litera * int(numar)
    return rezultat

print(comprima("aaabccdddd"), repr(comprima("")), decomprima("a3b1c2d4"), decomprima("x12y1"))
```

### G6. Top cuvinte
`top_cuvinte(text, k)` returnează lista primelor `k` perechi `(cuvânt, apariții)`, de la cel mai frecvent, iar la egalitate în ordine alfabetică. Fără `sorted` cu cheie: alegeți de `k` ori „cel mai bun” cuvânt rămas.

**Teste:** `("ana are mere ana are pere ana", 2)` → `[("ana", 3), ("are", 2)]` · `("b a c a b", 3)` → `[("a", 2), ("b", 2), ("c", 1)]`

```python
# G6. Top cuvinte
```

```python
# Soluție – G6
def top_cuvinte(text, k):
    frecventa = {}
    for c in text.lower().split():
        frecventa[c] = frecventa.get(c, 0) + 1
    rezultat = []
    for pas in range(k):
        cel_mai_bun = None
        for c, nr in frecventa.items():
            if cel_mai_bun is None or nr > frecventa[cel_mai_bun] or (nr == frecventa[cel_mai_bun] and c < cel_mai_bun):
                cel_mai_bun = c
        if cel_mai_bun is None:
            break
        rezultat.append((cel_mai_bun, frecventa[cel_mai_bun]))
        del frecventa[cel_mai_bun]
    return rezultat

print(top_cuvinte("ana are mere ana are pere ana", 2), top_cuvinte("b a c a b", 3))
```

### G7 (bonus). Două sume
`doua_sume(lista, tinta)` returnează pozițiile `(i, j)`, `i < j`, ale primei perechi cu suma `tinta`, sau `None`. Faceți-o cu **o singură parcurgere**, ținând într-un dicționar valorile văzute și pozițiile lor.

**Teste:** `([2, 7, 11, 15], 9)` → `(0, 1)` · `([3, 2, 4], 6)` → `(1, 2)` · `([1, 2], 7)` → `None`

```python
# G7 (bonus). Două sume
```

```python
# Soluție – G7 (bonus)
def doua_sume(lista, tinta):
    vazute = {}
    for j in range(len(lista)):
        complement = tinta - lista[j]
        if complement in vazute:
            return (vazute[complement], j)
        if lista[j] not in vazute:
            vazute[lista[j]] = j
    return None

print(doua_sume([2, 7, 11, 15], 9), doua_sume([3, 2, 4], 6), doua_sume([1, 2], 7))
```

### G8 (bonus). Paranteze corecte
`paranteze_corecte(text)` verifică dacă parantezele `()`, `[]`, `{}` sunt închise corect. Folosiți o listă ca **stivă**: la o paranteză deschisă faceți `append`, la una închisă verificați ultima deschisă cu `pop()`.

**Teste:** `"([]{})"` → `True` · `"([)]"` → `False` · `"(("` → `False` · `"a(b)c"` → `True` · `")("` → `False`

```python
# G8 (bonus). Paranteze corecte
```

```python
# Soluție – G8 (bonus)
def paranteze_corecte(text):
    pereche = {")": "(", "]": "[", "}": "{"}
    stiva = []
    for c in text:
        if c in "([{":
            stiva.append(c)
        elif c in ")]}":
            if len(stiva) == 0 or stiva.pop() != pereche[c]:
                return False
    return len(stiva) == 0

print(paranteze_corecte("([]{})"), paranteze_corecte("([)]"), paranteze_corecte("(("), paranteze_corecte("a(b)c"), paranteze_corecte(")("))
```

### G9 (bonus). Spargerea cifrului lui Cezar
Cifrul lui Cezar mută fiecare literă cu `pas` poziții în alfabet. Scrieți `cripteaza(text, pas)`, apoi `sparge(text_criptat)`, care încearcă toți cei 26 de pași și îl alege pe cel la care textul decriptat conține cele mai multe cuvinte din lista `["si", "de", "la", "este", "in", "nu", "un", "o", "care", "pe"]`.

**Test:** `sparge(cripteaza("pythonul este un limbaj care se invata usor si repede", 11))` → pasul `11` și textul original

```python
# G9 (bonus). Spargerea cifrului lui Cezar
```

```python
# Soluție – G9 (bonus)
ALFABET = "abcdefghijklmnopqrstuvwxyz"
CUVINTE = ["si", "de", "la", "este", "in", "nu", "un", "o", "care", "pe"]

def cripteaza(text, pas):
    rezultat = ""
    for c in text.lower():
        if c in ALFABET:
            rezultat += ALFABET[(ALFABET.index(c) + pas) % 26]
        else:
            rezultat += c
    return rezultat

def scor(text):
    s = 0
    for cuvant in text.split():
        if cuvant in CUVINTE:
            s += 1
    return s

def sparge(text_criptat):
    cel_mai_bun_pas = 0
    for pas in range(26):
        if scor(cripteaza(text_criptat, -pas)) > scor(cripteaza(text_criptat, -cel_mai_bun_pas)):
            cel_mai_bun_pas = pas
    return cel_mai_bun_pas, cripteaza(text_criptat, -cel_mai_bun_pas)

print(sparge(cripteaza("pythonul este un limbaj care se invata usor si repede", 11)))
```

### G10 (bonus). Cea mai lungă secvență crescătoare
`secventa_crescatoare(lista)` returnează cea mai lungă secvență de elemente **consecutive** strict crescătoare (prima, dacă sunt mai multe la fel de lungi).

**Teste:** `[1, 2, 0, 3, 4, 5, 1]` → `[0, 3, 4, 5]` · `[5, 4, 3]` → `[5]` · `[]` → `[]`

```python
# G10 (bonus). Cea mai lungă secvență crescătoare
```

```python
# Soluție – G10 (bonus)
def secventa_crescatoare(lista):
    if len(lista) == 0:
        return []
    start_bun, lungime_buna = 0, 1
    start = 0
    for i in range(1, len(lista) + 1):
        if i == len(lista) or lista[i] <= lista[i - 1]:
            if i - start > lungime_buna:
                start_bun, lungime_buna = start, i - start
            start = i
    return lista[start_bun:start_bun + lungime_buna]

print(secventa_crescatoare([1, 2, 0, 3, 4, 5, 1]), secventa_crescatoare([5, 4, 3]), secventa_crescatoare([]))
```

### ❓ Proiect A – Quiz

```python
# Referință – Quiz
import random

intrebari = [
    {"intrebare": "Câte zile are o săptămână?", "variante": ["5", "7", "10"], "corect": "7"},
    {"intrebare": "Care e cel mai mare ocean?", "variante": ["Atlantic", "Indian", "Pacific"], "corect": "Pacific"},
    {"intrebare": "Ce limbaj învățăm la acest modul?", "variante": ["Python", "Java", "C++"], "corect": "Python"},
    {"intrebare": "Cât face 7 * 8?", "variante": ["54", "56", "64"], "corect": "56"},
    {"intrebare": "Capitala Italiei?", "variante": ["Milano", "Roma", "Napoli"], "corect": "Roma"},
]

def normalizeaza(text):
    return text.strip().lower()

def verifica(intrebare, raspuns):
    return normalizeaza(raspuns) == normalizeaza(intrebare["corect"])

def mesaj_final(scor, total):
    procent = scor * 100 // total
    if procent == 100:
        return f"Perfect! {scor}/{total} 🏆"
    elif procent >= 60:
        return f"Foarte bine: {scor}/{total}"
    return f"{scor}/{total}. Mai încearcă!"

def o_runda():
    scor = 0
    random.shuffle(intrebari)
    for q in intrebari:
        print("\n" + q["intrebare"])
        for i, v in enumerate(q["variante"], start=1):
            print(f"  {i}) {v}")
        raspuns = input("Răspunsul tău (text sau număr): ")
        if raspuns.strip().isdigit() and 1 <= int(raspuns) <= len(q["variante"]):
            raspuns = q["variante"][int(raspuns) - 1]       # acceptă și numărul variantei
        if verifica(q, raspuns):
            print("✅ Corect!")
            scor += 1
        else:
            print(f"❌ Greșit, răspunsul era {q['corect']}.")
    print("\n" + mesaj_final(scor, len(intrebari)))
    return scor

record = 0
while True:
    scor = o_runda()
    record = max(record, scor)
    if input("Mai joci? (d/n): ").strip().lower() != "d":
        break
print(f"Record: {record}/{len(intrebari)}")
```

### 🪢 Proiect B – Spânzurătoarea

```python
# Referință – Spânzurătoarea
import random

cuvinte = ["python", "variabila", "functie", "lista", "dictionar", "algoritm",
           "calculator", "tastatura", "program", "internet"]
DESEN = ["😀", "🙂", "😐", "😕", "😟", "😨", "💀"]

def ascunde(cuvant, ghicite):
    return " ".join(litera if litera in ghicite else "_" for litera in cuvant)

def a_castigat(cuvant, ghicite):
    for litera in cuvant:
        if litera not in ghicite:
            return False
    return True

def citeste_litera(incercate):
    while True:
        litera = input("Literă: ").strip().lower()
        if len(litera) != 1 or not litera.isalpha():
            print("Scrie o singură literă.")
        elif litera in incercate:
            print("Ai mai încercat litera asta.")
        else:
            return litera

cuvant = random.choice(cuvinte)
ghicite = []
greseli = 0
MAX_GRESELI = 6

while greseli < MAX_GRESELI and not a_castigat(cuvant, ghicite):
    print(f"\n{DESEN[greseli]}  {ascunde(cuvant, ghicite)}   greșeli: {greseli}/{MAX_GRESELI}")
    litera = citeste_litera(ghicite)
    ghicite.append(litera)
    if litera not in cuvant:
        greseli += 1
        print("Nu e în cuvânt!")

if a_castigat(cuvant, ghicite):
    print(f"\n🎉 Ai câștigat! Cuvântul era {cuvant}.")
else:
    print(f"\n{DESEN[-1]} Ai pierdut. Cuvântul era {cuvant}.")
```

### 📊 Proiect C – Tracker de cheltuieli

```python
# Referință – Tracker de cheltuieli
cheltuieli = {"mancare": [], "transport": [], "distractie": []}

def adauga(cheltuieli, categorie, suma):
    if categorie not in cheltuieli:
        cheltuieli[categorie] = []
    cheltuieli[categorie].append(suma)

def total_categorie(cheltuieli, categorie):
    return sum(cheltuieli.get(categorie, []))

def total_general(cheltuieli):
    total = 0
    for categorie in cheltuieli:
        total += total_categorie(cheltuieli, categorie)
    return total

def categoria_maxima(cheltuieli):
    maxima = None
    for categorie in cheltuieli:
        if maxima is None or total_categorie(cheltuieli, categorie) > total_categorie(cheltuieli, maxima):
            maxima = categorie
    return maxima

def citeste_suma():
    while True:
        text = input("Suma (lei): ").replace(",", ".")
        try:
            suma = float(text)
            if suma > 0:
                return suma
        except ValueError:
            pass
        print("Scrie un număr pozitiv.")

BUGET = 200
while True:
    print("\n1. Adaugă cheltuială   2. Raport   0. Ieșire")
    optiune = input("Alege: ").strip()
    if optiune == "1":
        categorie = input("Categoria: ").strip().lower() or "altele"
        adauga(cheltuieli, categorie, citeste_suma())
    elif optiune == "2":
        total = total_general(cheltuieli)
        for categorie in cheltuieli:
            t = total_categorie(cheltuieli, categorie)
            procent = t * 100 / total if total else 0
            print(f"{categorie:12} {t:8.2f} lei  ({procent:.0f}%)")
        print(f"TOTAL: {total:.2f} lei · mai ai {BUGET - total:.2f} lei din buget")
        if total:
            print("Cele mai multe cheltuieli:", categoria_maxima(cheltuieli))
    elif optiune == "0":
        break
    else:
        print("Opțiune necunoscută.")
```

### 📒 Proiect D – Catalog

```python
# Referință – Catalog
catalog = {"Ana": [9, 10], "Mihai": [7, 8, 6]}

def adauga_nota(catalog, nume, nota):
    if nume not in catalog:
        catalog[nume] = []
    catalog[nume].append(nota)

def media(note):
    if len(note) == 0:
        return 0
    return sum(note) / len(note)

def clasament(catalog):
    perechi = []
    for nume, note in catalog.items():
        perechi.append((media(note), nume))
    return sorted(perechi, reverse=True)

def citeste_nota():
    while True:
        text = input("Nota (1-10): ").strip()
        if text.isdigit() and 1 <= int(text) <= 10:
            return int(text)
        print("Notă invalidă.")

while True:
    print("\n1. Adaugă notă   2. Mediile   3. Clasament   0. Ieșire")
    optiune = input("Alege: ").strip()
    if optiune == "1":
        nume = input("Elev: ").strip().title()
        if nume:
            adauga_nota(catalog, nume, citeste_nota())
    elif optiune == "2":
        for nume, note in catalog.items():
            print(f"{nume}: {note} → media {media(note):.2f}")
    elif optiune == "3":
        for loc, (m, nume) in enumerate(clasament(catalog), start=1):
            medalie = {1: "🥇", 2: "🥈", 3: "🥉"}.get(loc, "  ")
            print(f"{medalie} {loc}. {nume} – {m:.2f}")
    elif optiune == "0":
        break
    else:
        print("Opțiune necunoscută.")
```

### 🤖 Proiect E – Chatbot bazat pe reguli

```python
# Referință – Chatbot
import random

reguli = {
    "salut": ["Salut! 👋", "Bună! Ce mai faci?"],
    "ce faci": ["Învăț Python, ca și tine.", "Răspund la întrebări. Tu?"],
    "nume": ["Mă numesc PyBot."],
    "python": ["Python e limbajul meu preferat!", "Știai că Python a fost numit după un show de comedie?"],
    "scoala": ["Școala e mai faină cu programare. 😄"],
    "vreme": ["Nu am fereastră, dar sper că e soare."],
    "gluma": ["De ce s-a dus programatorul la doctor? Avea un bug. 🐛"],
    "multumesc": ["Cu plăcere!"],
}
NU_STIU = ["Nu am înțeles. Poți reformula?", "Hmm, despre asta nu știu încă."]
memorie = {"nume": None, "neintelese": 0}

def extrage_nume(mesaj):
    cuvinte = mesaj.split()
    for i in range(len(cuvinte) - 1):
        if cuvinte[i].lower() in ("numesc", "sunt"):
            return cuvinte[i + 1].strip(".,!").title()
    return None

def raspunde(mesaj):
    nume = extrage_nume(mesaj)
    if nume:
        memorie["nume"] = nume
        return f"Încântat, {nume}!"
    text = mesaj.lower()
    for cuvant_cheie in reguli:
        if cuvant_cheie in text:
            raspuns = random.choice(reguli[cuvant_cheie])
            if memorie["nume"]:
                raspuns = f"{memorie['nume']}, {raspuns[0].lower()}{raspuns[1:]}"
            return raspuns
    memorie["neintelese"] += 1
    return random.choice(NU_STIU)

print("PyBot: Salut! Scrie „pa” ca să ieși.")
while True:
    mesaj = input("Tu: ")
    if mesaj.strip().lower() == "pa":
        print(f"PyBot: La revedere! (n-am înțeles {memorie['neintelese']} mesaje 😅)")
        break
    print("PyBot:", raspunde(mesaj))
```

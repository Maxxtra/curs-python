## 1. Algoritm și program

## 4. Operatori – recapitulare cu capcane

| | |
|---|---|
| aritmetici | `+ - * / // % **` |
| comparație | `== != < > <= >=`, se pot înlănțui: `0 <= x < 10` |
| logici | `and or not` (se evaluează de la stânga, se opresc când știu răspunsul) |

**`//` și `%` pentru numere negative:** Python rotunjește câtul **în jos** (spre minus infinit), iar restul are semnul împărțitorului.

````python
print(7 // 2, 7 % 2)
print(-7 // 2, -7 % 2)       # -4 și 1, nu -3 și -1!
print(2 ** 3 ** 2)           # ** se evaluează de la dreapta: 2 ** 9
print(-2 ** 2)               # - se aplică după **
print(10 - 4 - 3, 2 * 3 % 4)
x = 7
print(0 < x < 10, 5 < x == 7)
print(not x > 5 or x < 0)
````

> 🧑‍🏫 **Note pentru profesor**

> Răspunsuri: `3 1` · `-4 1` · `512` · `-4` · `3 2` · `True True` · `False`.

## 5. Decizii: `if`, `elif`, `else`

Două lucruri care fac diferența la problemele grele:
1. **Ordinea ramurilor** contează: Python se oprește la prima condiție adevărată. Pune cazurile particulare **înaintea** celor generale.
2. **`if`-uri separate** vs **`elif`**: cu `if` separate se verifică toate; cu `elif` se intră în cel mult una.

````python
n = 15
if n % 3 == 0:
    print("Fizz")
elif n % 15 == 0:          # nu se ajunge niciodată aici pentru n = 15. De ce?
    print("FizzBuzz")

if n % 3 == 0:
    print("divizibil cu 3")
if n % 5 == 0:
    print("divizibil cu 5")
````

### `if` în `if` vs condiții combinate

````python
varsta = 16
are_bilet = True

if varsta >= 14:
    if are_bilet:
        print("Intră")

if varsta >= 14 and are_bilet:     # echivalent, mai ușor de citit
    print("Intră")
````

### C2. Triunghiul complet
Citește trei laturi. Afișează dacă formează un triunghi și, dacă da: tipul după laturi (echilateral / isoscel / oarecare), dacă e **dreptunghic** (Pitagora; atenție, laturile pot fi în orice ordine și pot fi `float`) și **aria** cu formula lui Heron: `s = (a+b+c)/2`, `A = (s(s-a)(s-b)(s-c)) ** 0.5`.

**Teste:** `3 4 5` → oarecare, dreptunghic, aria 6.00 · `5 3 4` → la fel · `2 2 2` → echilateral, aria 1.73 · `1 2 3` → nu e triunghi

````python
# C2. Triunghiul complet
````

````python
# Soluție – C2
a = float(input("a = "))
b = float(input("b = "))
c = float(input("c = "))

if a + b <= c or a + c <= b or b + c <= a:
    print("Nu se poate forma un triunghi")
else:
    if a == b == c:
        print("Echilateral")
    elif a == b or b == c or a == c:
        print("Isoscel")
    else:
        print("Oarecare")
    # cea mai mare latură e ipotenuza candidată
    if a >= b and a >= c:
        ip, c1, c2 = a, b, c
    elif b >= a and b >= c:
        ip, c1, c2 = b, a, c
    else:
        ip, c1, c2 = c, a, b
    if abs(c1 ** 2 + c2 ** 2 - ip ** 2) < 1e-9:
        print("Dreptunghic")
    s = (a + b + c) / 2
    aria = (s * (s - a) * (s - b) * (s - c)) ** 0.5
    print(f"Aria: {aria:.2f}")
````

### C3. În ce zi a săptămânii?
Pentru o dată (zi, lună, an), ziua săptămânii se calculează cu **formula lui Zeller**: dacă luna e ianuarie sau februarie, o tratezi ca luna 13 sau 14 a anului **precedent**. Apoi, cu `K = an % 100` și `J = an // 100`:

`h = (zi + (13 * (luna + 1)) // 5 + K + K // 4 + J // 4 + 5 * J) % 7`

`h = 0` înseamnă sâmbătă, `1` duminică, `2` luni … `6` vineri. Afișează numele zilei.

**Teste:** `14 10 2026` → miercuri · `1 1 2000` → sâmbătă · `25 12 2026` → vineri · `29 2 2024` → joi

````python
# C3. În ce zi a săptămânii?
````

````python
# Soluție – C3
zi = int(input("Ziua: "))
luna = int(input("Luna: "))
an = int(input("Anul: "))
if luna <= 2:
    luna += 12
    an -= 1
K = an % 100
J = an // 100
h = (zi + (13 * (luna + 1)) // 5 + K + K // 4 + J // 4 + 5 * J) % 7
if h == 0:
    print("sâmbătă")
elif h == 1:
    print("duminică")
elif h == 2:
    print("luni")
elif h == 3:
    print("marți")
elif h == 4:
    print("miercuri")
elif h == 5:
    print("joi")
else:
    print("vineri")
````

````python
# Soluție – C5
import random

alegere = input("Alege (piatra, spock, hartie, sopirla, foarfece): ").strip().lower()
if alegere == "piatra":
    om = 0
elif alegere == "spock":
    om = 1
elif alegere == "hartie":
    om = 2
elif alegere == "sopirla":
    om = 3
elif alegere == "foarfece":
    om = 4
else:
    om = -1

if om == -1:
    print("Alegere invalidă")
else:
    calc = random.randint(0, 4)
    if calc == 0:
        print("Calculatorul a ales: piatra")
    elif calc == 1:
        print("Calculatorul a ales: spock")
    elif calc == 2:
        print("Calculatorul a ales: hartie")
    elif calc == 3:
        print("Calculatorul a ales: sopirla")
    else:
        print("Calculatorul a ales: foarfece")
    d = (om - calc) % 5
    if d == 0:
        print("Egalitate")
    elif d == 1 or d == 2:
        print("Ai câștigat!")
    else:
        print("A câștigat calculatorul")
````

Reduceri: cod `ELEV10` dă 10%; peste 500 lei se adaugă încă 5% (din prețul deja redus); reducerea totală nu poate depăși 100 lei. Citește `P` și codul și afișează un bon frumos: preț, reducere, livrare, total, aliniat la dreapta.

````python
# Soluție – C6 (bonus)
P = float(input("Prețul produselor: "))
cod = input("Cod reducere (sau -): ").strip().upper()

redus = P
if cod == "ELEV10":
    redus = redus * 0.90
if P > 500:
    redus = redus * 0.95
reducere = P - redus
if reducere > 100:
    reducere = 100
subtotal = P - reducere
if subtotal >= 200:
    livrare = 0
else:
    livrare = 20
total = subtotal + livrare

print(f"{'Produse:':<12}{P:>10.2f}")
print(f"{'Reducere:':<12}{-reducere:>10.2f}")
print(f"{'Livrare:':<12}{livrare:>10.2f}")
print(f"{'TOTAL:':<12}{total:>10.2f}")
````

## 8. Mini-proiect: Calculatorul avansat 🧮

În pereche, construiți un calculator care cere două numere și o operație.

**Obligatoriu:**
1. Operațiile `+ - * / // % **`.
2. Împărțirea la zero (pentru `/`, `//` și `%`) dă un mesaj, nu o eroare.
3. Operator necunoscut → mesaj.
4. Dacă rezultatul e întreg (de exemplu `6 / 3`), se afișează **fără** `.0`; altfel cu cel mult 4 zecimale.
5. Numerele introduse greșit nu trebuie să blocheze programul: verificați **înainte** de conversie că textul arată ca un număr (indiciu: `text.replace(".", "", 1).replace("-", "", 1).isdigit()`).

**Extra:**
- ⭐ operatorul `r` (radical de ordin `b` din `a`: `a ** (1 / b)`), cu mesaj pentru radical de ordin par din număr negativ;
- ⭐ operatorul `cmp`, care spune care număr e mai mare și cu cât la sută;
- ⭐⭐ un mod „convertor”: grade C ↔ F, km ↔ mile, lei ↔ euro (cu un curs pe care îl citiți);
- ⭐⭐⭐ ideea voastră.

**La final** fiecare pereche prezintă: o operație normală, două cazuri „rele” și un extra.

````python
# Referință – Calculatorul avansat
text_a = input("Primul număr: ").strip()
op = input("Operația (+ - * / // % ** r cmp): ").strip()
text_b = input("Al doilea număr: ").strip()

ok_a = text_a.replace(".", "", 1).replace("-", "", 1).isdigit()
ok_b = text_b.replace(".", "", 1).replace("-", "", 1).isdigit()

if not ok_a or not ok_b:
    print("Nu ați introdus numere valide.")
else:
    a = float(text_a)
    b = float(text_b)
    rezultat = None
    if op == "+":
        rezultat = a + b
    elif op == "-":
        rezultat = a - b
    elif op == "*":
        rezultat = a * b
    elif op == "/" or op == "//" or op == "%":
        if b == 0:
            print("Nu se poate împărți la zero!")
        elif op == "/":
            rezultat = a / b
        elif op == "//":
            rezultat = a // b
        else:
            rezultat = a % b
    elif op == "**":
        rezultat = a ** b
    elif op == "r":
        if b == 0:
            print("Radical de ordin 0 nu există.")
        elif a < 0 and b % 2 == 0:
            print("Radical de ordin par din număr negativ nu e real.")
        elif a < 0:
            rezultat = -((-a) ** (1 / b))
        else:
            rezultat = a ** (1 / b)
    elif op == "cmp":
        if a == b:
            print("Numerele sunt egale.")
        else:
            if a > b:
                mare, mic = a, b
            else:
                mare, mic = b, a
            if mic == 0:
                print(f"{mare:g} e mai mare decât {mic:g}")
            else:
                print(f"{mare:g} e mai mare decât {mic:g} cu {(mare - mic) / abs(mic):.1%}")
    else:
        print("Operator necunoscut")

    if rezultat is not None:
        if rezultat == int(rezultat):
            print(f"{a:g} {op} {b:g} = {int(rezultat)}")
        else:
            print(f"{a:g} {op} {b:g} = {round(rezultat, 4)}")
````

> 🧑‍🏫 **Note pentru profesor**
>
> Punctul 5 (validarea fără erori) e cel care separă perechile: majoritatea vor încerca `int()` direct și vor primi ValueError. Lasă-i să ajungă singuri la ideea „verific înainte să convertesc”.
> 
> `:g` în f-string afișează numărul fără zerouri inutile (`6.0` → `6`). Dacă îl descoperă cineva, arată-l clasei.

## 0. Încălzire
Ce afișează? Discutați în bancă, apoi rulați.

````python
def puncte(n):
    total = 0
    for i in range(1, n + 1):
        if i % 2 == 0:
            total += i
    return total

print(puncte(6))
print(puncte(5) * 2)
````

> 🧑‍🏫 **Note pentru profesor**
>
> Răspuns: `puncte(6)` = 2+4+6 = **12**; `puncte(5)` = 2+4 = 6, deci **12**. Folosește-o ca să vezi rapid nivelul clasei (cine urmărește bucla și `return`). Apoi întrebarea care pornește ziua: „vreau să țin notele a 28 de elevi; fac 28 de variabile?”

### Funcții care primesc și returnează liste
Tiparul pe care îl veți folosi în toate misiunile: pornești de la o listă **goală**, parcurgi lista primită, adaugi ce te interesează, **returnezi** lista nouă.

````python
def peste_medie(note):
    media = sum(note) / len(note)
    rezultat = []
    for nota in note:
        if nota > media:
            rezultat.append(nota)
    return rezultat

print(peste_medie([9, 7, 10, 8, 6]))    # media 8 → [9, 10]
````

> 🧑‍🏫 **Note pentru profesor**
>
> Greșeli tipice: `note.append = 10`; `return` pus în interiorul buclei (funcția se oprește după primul element); lista rezultat creată în buclă, deci resetată la fiecare pas. Arată-l pe ultimul intenționat, le prinde bine la misiuni.

````python
# ⚙️ Verificările – rulați o dată, la început
_LIT = {}
_TESTE = {
    1: ("media", [(([9, 7, 10, 8, 6],), 8.0), (([10],), 10.0), (([5, 6],), 5.5)]),
    2: ("numar_pozitive", [(([3, -1, 0, 7, -5],), 2), (([-2, -3],), 0), (([],), 0)]),
    3: ("fara_duplicate", [(([1, 2, 2, 3, 1],), [1, 2, 3]), ((["a", "b", "a"],), ["a", "b"]), (([],), [])]),
    4: ("numar_vocale", [(("Informatica",), 5), (("PYTHON",), 1), (("xyz",), 0), (("aeiou",), 5)]),
    5: ("initiale", [(("ana maria popescu",), "A.M.P."), (("Ion",), "I."), (("radu ionescu",), "R.I.")]),
    6: ("este_palindrom", [(("cojoc",), True), (("Ele fac cafele",), True), (("python",), False)]),
    7: ("cuvinte_lungi", [(("ana are mere foarte gustoase", 4), ["foarte", "gustoase"]), (("un an", 5), []), (("calculator rapid", 5), ["calculator"])]),
    8: ("pozitia_maximului", [(([3, 9, 2, 9],), 1), (([5],), 0), (([-4, -1, -7],), 1)]),
}
_LITERE = "ALGORITM"

def verifica(misiune):
    nume, teste = _TESTE[misiune]
    f = globals().get(nume)
    if f is None:
        print(f"❌ Nu găsesc funcția {nume}. Ați rulat celula în care o scrieți? Ați scris numele exact?")
        return
    for argumente, asteptat in teste:
        apel = f"{nume}({', '.join(repr(a) for a in argumente)})"
        try:
            primit = f(*[list(a) if isinstance(a, list) else a for a in argumente])
        except Exception as e:
            print(f"❌ {apel} a dat eroare: {type(e).__name__}: {e}")
            return
        if isinstance(asteptat, float) and isinstance(primit, (int, float)):
            corect = abs(primit - asteptat) < 1e-9
        else:
            corect = primit == asteptat
        if not corect:
            print(f"❌ {apel} a dat {primit!r}, dar trebuia {asteptat!r}.")
            if primit is None:
                print("   Indiciu: lipsește return?")
            return
    _LIT[misiune] = _LITERE[misiune - 1]
    print(f"✅ Misiunea {misiune} e corectă! Litera voastră: „{_LITERE[misiune - 1]}”. Schimbați rolurile!")

def cuvant_secret():
    c = "".join(_LIT.get(i, "_") for i in range(1, 9))
    print("Cuvântul vostru:", " ".join(c))
    print("🎉 GATA! Scrieți-vă numele pe tablă." if "_" not in c else "Locurile cu _ sunt misiunile care lipsesc.")

print("Verificările sunt gata. Începeți cu Misiunea 1!")
````

### Misiunea 3 – Fără duplicate
Scrieți `fara_duplicate(lista)` care returnează o listă **nouă**, fără elementele repetate, păstrând ordinea primei apariții.

Exemplu: `fara_duplicate([1, 2, 2, 3, 1])` → `[1, 2, 3]`

````python
def fara_duplicate(lista):
    rezultat = []
    for x in lista:
        if x not in rezultat:
            rezultat.append(x)
    return rezultat
````

Exemplu: `pozitia_maximului([3, 9, 2, 9])` → `1`

````python
def pozitia_maximului(lista):
    poz = 0
    for i in range(1, len(lista)):
        if lista[i] > lista[poz]:
            poz = i
    return poz
````

## 4. Dicționare

O listă găsește valorile după **poziție**. Un **dicționar** le găsește după **cheie**, ca un dicționar adevărat: cauți cuvântul, găsești explicația.

````python
elev = {"nume": "Ana", "varsta": 16, "clasa": "X A"}
print(elev["nume"])

elev["varsta"] = 17            # schimbăm
elev["oras"] = "București"     # adăugăm
print(elev)

print("oras" in elev)          # există cheia?
print(elev.get("telefon", "necunoscut"))   # fără KeyError dacă lipsește

for cheie, valoare in elev.items():
    print(cheie, "→", valoare)
````

| Listă | Dicționar |
|---|---|
| `[9, 7, 10]` | `{"Ana": 9, "Mihai": 7}` |
| după poziție: `note[0]` | după cheie: `note["Ana"]` |
| `append` adaugă | `d[cheie] = valoare` adaugă sau schimbă |

### Tiparul „numărăm cu un dicționar”

````python
text = "abracadabra"
frecventa = {}
for litera in text:
    frecventa[litera] = frecventa.get(litera, 0) + 1
print(frecventa)
````

> 🧑‍🏫 **Note pentru profesor**
>
> Desenează la tablă cum crește dicționarul pentru primele 4 litere: `{a:1}` → `{a:1, b:1}` → `{a:1, b:1, r:1}` → `{a:2, b:1, r:1}`. Cheia înțelegerii: `get(litera, 0)` = „cât am numărat până acum, sau 0 dacă e prima dată”.

### Sortăm după valoare: cele mai frecvente

````python
perechi = sorted(frecventa.items(), key=lambda p: p[1], reverse=True)
print(perechi)
for litera, nr in perechi[:3]:
    print(litera, nr)
````

`key=lambda p: p[1]` înseamnă „sortează perechile după al doilea element”, adică după frecvență. E singura bucată nouă de sintaxă; o puteți folosi ca atare.

### Listă de dicționare = un mic tabel

````python
catalog = [
    {"nume": "Ana", "nota": 9},
    {"nume": "Mihai", "nota": 7},
    {"nume": "Ioana", "nota": 10},
]
for elev in catalog:
    print(f"{elev['nume']}: {elev['nota']}")

cel_mai_bun = max(catalog, key=lambda e: e["nota"])
print("Cel mai bun:", cel_mai_bun["nume"])
````

## 5. Mini-proiect: Analizorul de text 🔍

Construiți un program care primește un text și afișează un mic raport, ca un instrument adevărat de analiză. Lucrați în perechi și **refolosiți funcțiile din misiuni**.

**Raportul trebuie să conțină:**
1. numărul de cuvinte și numărul de caractere (fără spații);
2. cel mai lung cuvânt;
3. numărul de vocale;
4. **top 5 cele mai frecvente cuvinte**.

**Extra:** ignorați cuvintele scurte (`și`, `de`, `la`…) · comparați două texte (care are cuvinte mai lungi?) · rulați analizorul pe versurile unei melodii sau pe un articol.

````python
# Referință – Analizorul de text
TEXT = """Python este un limbaj de programare. Python este simplu si Python este folosit
in inteligenta artificiala, in analiza datelor si in aplicatii web."""
CUVINTE_SCURTE = {"si", "de", "in", "un", "o", "la", "cu", "pe"}

def curata(text):
    for semn in ".,!?;:\"()":
        text = text.replace(semn, "")
    return text.lower()

def cel_mai_lung_cuvant(cuvinte):
    cel_mai_lung = ""
    for c in cuvinte:
        if len(c) > len(cel_mai_lung):
            cel_mai_lung = c
    return cel_mai_lung

def top_cuvinte(cuvinte, k, ignora=()):
    frecventa = {}
    for c in cuvinte:
        if c not in ignora:
            frecventa[c] = frecventa.get(c, 0) + 1
    return sorted(frecventa.items(), key=lambda p: p[1], reverse=True)[:k]

cuvinte = curata(TEXT).split()
print("Cuvinte:", len(cuvinte))
print("Caractere fără spații:", len(TEXT.replace(" ", "").replace("\n", "")))
print("Cel mai lung cuvânt:", cel_mai_lung_cuvant(cuvinte))
print("Vocale:", numar_vocale(TEXT))
print("Top 5 (fără cuvinte scurte):")
for cuvant, nr in top_cuvinte(cuvinte, 5, CUVINTE_SCURTE):
    print(f"  {cuvant}: {nr}")
````

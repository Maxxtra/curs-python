### ⚙️ Rulează mai întâi celula de mai jos (pregătește verificările)

````python
# ⚙️ Pregătirea verificărilor – rulează o dată, la început
def verifica_valoare(nume, valoare, corect):
    if valoare == corect:
        print(f"✅ Corect! {nume} = {valoare}")
    else:
        print(f"❌ {nume} este {valoare}, dar trebuia să fie altceva. Mai încearcă!")

def verifica_functie(f, teste):
    for argumente, asteptat in teste:
        primit = f(*argumente)
        apel = f"{f.__name__}{argumente}".replace(",)", ")")
        if primit != asteptat:
            print(f"❌ {apel} a dat {primit}, dar trebuia {asteptat}.")
            if primit is None:
                print("   Indiciu: ai uitat return?")
            return
    print(f"✅ Funcția {f.__name__} merge pentru toate testele!")

print("Gata, putem începe!")
````

## 1. Bucla `while` – „repetă cât timp…” ⏱ 09:45–10:20

Data trecută, `if` făcea ceva **o singură dată**, dacă o condiție era adevărată.
`while` face ceva **din nou și din nou**, **cât timp** condiția e adevărată.

````python
i = 1
while i <= 5:
    print(i)
    i = i + 1
print("Gata!")
````

### Ce s-a întâmplat, pas cu pas

| Pas | `i` | `i <= 5`? | Ce face |
|---|---|---|---|
| 1 | 1 | True | afișează 1, `i` devine 2 |
| 2 | 2 | True | afișează 2, `i` devine 3 |
| 3 | 3 | True | afișează 3, `i` devine 4 |
| 4 | 4 | True | afișează 4, `i` devine 5 |
| 5 | 5 | True | afișează 5, `i` devine 6 |
| 6 | 6 | **False** | **se oprește**, afișează „Gata!” |

Orice `while` are nevoie de 3 lucruri:
1. **start**: `i = 1`
2. **condiție**: `i <= 5`
3. **pas**: `i = i + 1` ← fără el, bucla nu se oprește niciodată!

> 🧑‍🏫 **Note pentru profesor**
>
> Desenează tabelul la tablă și completați-l împreună, rând cu rând, **înainte** să arăți varianta din notebook. Pune-i să spună cu voce tare ce se întâmplă la pasul 6. Asta e cheia: bucla se oprește când condiția devine `False`.

````python
secunde = 5
while secunde > 0:
    print(secunde)
    secunde = secunde - 1
print("Lansare! 🚀")
````

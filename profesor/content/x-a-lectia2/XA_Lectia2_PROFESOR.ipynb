### Algoritmul lui Euclid

````python
a, b = 1071, 462
while b != 0:
    a, b = b, a % b
    print(a, b)
print("cmmdc =", a)
````

### Căutare prin înjumătățire
Calculatorul ghicește un număr ascuns între 1 și 1000 întrebând doar „mai mic sau mai mare?”. Câte întrebări îi trebuie în cel mai rău caz?

````python
secret = 737
stanga, dreapta = 1, 1000
pasi = 0
while stanga <= dreapta:
    mijloc = (stanga + dreapta) // 2
    pasi += 1
    if mijloc == secret:
        break
    elif mijloc < secret:
        stanga = mijloc + 1
    else:
        dreapta = mijloc - 1
print(f"Găsit {mijloc} în {pasi} pași")
````

### `break` și `else` pe buclă: căutăm ceva

````python
n = 91
for d in range(2, n):
    if n % d == 0:
        print(f"{n} nu e prim: se împarte la {d}")
        break
else:
    print(f"{n} e prim")     # rulează doar dacă bucla NU s-a oprit cu break
````

`else` după `for` pare ciudat, dar e exact „am căutat peste tot și n-am găsit”. Nu e obligatoriu: același lucru se face cu o variabilă `gasit = False`.

### Cât de repede?
Pentru `n` prim, bucla de mai sus face aproape `n` pași. Dar dacă `n = a × b`, unul dintre `a`, `b` e cel mult `√n`. Deci e suficient să verificăm divizorii până la `√n`:

````python
n = 1_000_003
pasi = 0
d = 2
prim = True
while d * d <= n:          # echivalent cu d <= radical din n, fără float
    pasi += 1
    if n % d == 0:
        prim = False
        break
    d += 1
if prim:
    print(n, "prim | pași:", pasi)
else:
    print(n, "compus | pași:", pasi)
````

## 3. Funcții

Funcțiile sunt cărămizile oricărui program mai mare: **o funcție = o singură treabă**, cu un nume clar, care **returnează** rezultatul (nu îl afișează).

````python
def suma_cifre(n):
    s = 0
    while n > 0:
        s += n % 10
        n //= 10
    return s

def este_prim(n):
    if n < 2:
        return False
    d = 2
    while d * d <= n:
        if n % d == 0:
            return False
        d += 1
    return True

# funcțiile se combină: numerele prime de 2 cifre cu suma cifrelor tot primă
for x in range(10, 100):
    if este_prim(x) and este_prim(suma_cifre(x)):
        print(x, end=" ")
print()
````

🔹 **Încearcă tu:**
- **F1.** `oglindit(n)` returnează numărul cu cifrele în ordine inversă: `oglindit(1230)` → `321`.
- **F2.** `este_palindrom(n)` folosind `oglindit`.
- **F3.** `numar_cifre(n, cifra)` returnează de câte ori apare `cifra` în `n`: `numar_cifre(1223242, 2)` → `4`.

````python
# F1, F2, F3
````

````python
# Soluție – F1–F3
def oglindit(n):
    r = 0
    while n > 0:
        r = r * 10 + n % 10
        n //= 10
    return r

def este_palindrom(n):
    return n == oglindit(n)

def numar_cifre(n, cifra):
    c = 0
    while n > 0:
        if n % 10 == cifra:
            c += 1
        n //= 10
    return c

print(oglindit(1230), este_palindrom(12321), este_palindrom(1230), numar_cifre(1223242, 2))
````

### E2. Numere perfecte
Un număr e **perfect** dacă e egal cu suma divizorilor lui mai mici decât el (6 = 1 + 2 + 3). Scrie `suma_divizori(n)` și afișează toate numerele perfecte până la 10 000.

**Teste:** `suma_divizori(12)` → `16` (1+2+3+4+6) · perfecte până la 10 000: `6 28 496 8128`

⭐ Fă `suma_divizori` să meargă doar până la `√n` (divizorii vin în perechi `d` și `n // d`).

````python
# E2. Numere perfecte
````

````python
# Soluție – E2
def suma_divizori(n):
    if n == 1:
        return 0
    s = 1
    d = 2
    while d * d <= n:
        if n % d == 0:
            s += d
            if d != n // d:
                s += n // d
        d += 1
    return s

print(suma_divizori(12))
for x in range(2, 10001):
    if suma_divizori(x) == x:
        print(x, end=" ")
print()
````

### E4. Fracții
Scrie `cmmdc(a, b)` (Euclid), `cmmmc(a, b)` (indiciu: `a * b // cmmdc(a, b)`) și `simplifica(a, b)` care returnează textul fracției ireductibile, de exemplu `"3/4"`. Dacă numitorul devine 1, returnează doar numărătorul.

**Teste:** `cmmdc(84, 36)` → `12` · `cmmmc(4, 6)` → `12` · `simplifica(84, 36)` → `"7/3"` · `simplifica(10, 5)` → `"2"` · `simplifica(3, 7)` → `"3/7"`

⭐ Adună două fracții: `aduna(1, 4, 1, 6)` → `"5/12"`.

````python
# E4. Fracții
````

````python
# Soluție – E4
def cmmdc(a, b):
    while b != 0:
        a, b = b, a % b
    return a

def cmmmc(a, b):
    return a * b // cmmdc(a, b)

def simplifica(a, b):
    d = cmmdc(a, b)
    a, b = a // d, b // d
    if b == 1:
        return str(a)
    return f"{a}/{b}"

def aduna(a, b, c, d):
    numitor = cmmmc(b, d)
    return simplifica(a * (numitor // b) + c * (numitor // d), numitor)

print(cmmdc(84, 36), cmmmc(4, 6), simplifica(84, 36), simplifica(10, 5), simplifica(3, 7), aduna(1, 4, 1, 6))
````

### E5. Baza 2
Scrie `in_baza2(n)`, care returnează scrierea lui `n` în baza 2 ca text (împarți repetat la 2 și citești resturile **de jos în sus**), și `din_baza2(text)`, care face invers (pentru fiecare cifră, de la stânga: `rezultat = rezultat * 2 + cifra`).

**Teste:** `in_baza2(13)` → `"1101"` · `in_baza2(0)` → `"0"` · `in_baza2(1024)` → `"10000000000"` · `din_baza2("1101")` → `13` · `din_baza2(in_baza2(2026))` → `2026`

⭐ Generalizează: `in_baza(n, b)` pentru orice bază de la 2 la 16 (cifrele peste 9 sunt `A`–`F`): `in_baza(255, 16)` → `"FF"`.

````python
# E5. Baza 2
````

````python
# Soluție – E5
def in_baza2(n):
    if n == 0:
        return "0"
    rezultat = ""
    while n > 0:
        rezultat = str(n % 2) + rezultat
        n //= 2
    return rezultat

def din_baza2(text):
    rezultat = 0
    for c in text:
        rezultat = rezultat * 2 + int(c)
    return rezultat

def cifra_in_text(r):
    k = 0
    for c in "0123456789ABCDEF":
        if k == r:
            return c
        k += 1

def in_baza(n, b):
    if n == 0:
        return "0"
    rezultat = ""
    while n > 0:
        rezultat = cifra_in_text(n % b) + rezultat
        n //= b
    return rezultat

print(in_baza2(13), in_baza2(0), in_baza2(1024), din_baza2("1101"), din_baza2(in_baza2(2026)), in_baza(255, 16))
````

### E8 (bonus). Numere Armstrong
Un număr cu `k` cifre e **Armstrong** dacă e egal cu suma cifrelor ridicate la puterea `k`: 153 = 1³ + 5³ + 3³. Scrie `este_armstrong(n)` și afișează toate numerele Armstrong mai mici decât 100 000.

**Teste:** `1 2 3 4 5 6 7 8 9 153 370 371 407 1634 8208 9474 54748 92727 93084`

````python
# E8 (bonus). Numere Armstrong
````

````python
# Soluție – E8 (bonus)
def numar_cifre(n):
    k = 0
    while n > 0:
        k += 1
        n //= 10
    return k

def este_armstrong(n):
    k = numar_cifre(n)
    s = 0
    m = n
    while m > 0:
        s += (m % 10) ** k
        m //= 10
    return s == n

for x in range(1, 100000):
    if este_armstrong(x):
        print(x, end=" ")
print()
````

### E9 (bonus). Cel mai lung lanț Collatz
Din `n`: dacă e par devine `n // 2`, altfel `3 * n + 1`, până la 1. Scrie `lungime_collatz(n)` (numărul de pași) și află ce număr sub 10 000 are cel mai lung lanț.

**Teste:** `lungime_collatz(1)` → `0` · `lungime_collatz(6)` → `8` · sub 10 000: `6171`, cu `261` pași

````python
# E9 (bonus). Cel mai lung lanț Collatz
````

````python
# Soluție – E9 (bonus)
def lungime_collatz(n):
    pasi = 0
    while n != 1:
        if n % 2 == 0:
            n //= 2
        else:
            n = 3 * n + 1
        pasi += 1
    return pasi

print(lungime_collatz(1), lungime_collatz(6))
cel_mai_bun, maxim = 1, 0
for x in range(1, 10000):
    L = lungime_collatz(x)
    if L > maxim:
        cel_mai_bun, maxim = x, L
print(cel_mai_bun, maxim)
````

### B1. Prim sau nu?
Ar trebui să spună corect dacă un număr e prim.

````python
def este_prim(n):
    if n < 2:
        return False
    for d in range(2, n):
        if n % d == 0:
            return False
        else:
            return True

print(este_prim(7), este_prim(9), este_prim(2))
````

**Bug:** `return True` e în `else`, deci decide după primul divizor (d = 2): `este_prim(9)` dă `True`, iar `este_prim(2)` dă `None` (bucla nu rulează). `return True` trebuie mutat după buclă.

````python
# Soluție – B1
def este_prim(n):
    if n < 2:
        return False
    for d in range(2, n):
        if n % d == 0:
            return False
    return True

print(este_prim(7), este_prim(9), este_prim(2))
````

### B7. Numărul de cifre
`numar_cifre(0)` ar trebui să dea `1`, iar `numar_cifre(-305)` să dea `3`.

````python
def numar_cifre(n):
    k = 0
    while n > 0:
        k += 1
        n //= 10
    return k

print(numar_cifre(2026), numar_cifre(0), numar_cifre(-305))
````

**Bug:** Pentru 0 bucla nu rulează (dă 0), iar pentru numere negative la fel. Reparație: `n = abs(n)` la început și caz special pentru 0 (sau o buclă care rulează măcar o dată).

````python
# Soluție – B7
def numar_cifre(n):
    n = abs(n)
    k = 1
    while n >= 10:
        k += 1
        n //= 10
    return k

print(numar_cifre(2026), numar_cifre(0), numar_cifre(-305))
````

### 🎯 Calculatorul ghicește

````python
# Referință – Calculatorul ghicește
def citeste_raspuns(mijloc):
    while True:
        r = input(f"Este {mijloc}? (mic / mare / da): ").strip().lower()
        if r == "mic" or r == "mare" or r == "da":
            return r
        print("Răspunde cu mic, mare sau da.")

def joaca(minim=1, maxim=1000):
    print(f"Gândește-te la un număr între {minim} și {maxim}.")
    stanga, dreapta = minim, maxim
    intrebari = 0
    while stanga <= dreapta:
        mijloc = (stanga + dreapta) // 2
        intrebari += 1
        r = citeste_raspuns(mijloc)
        if r == "da":
            print(f"Am ghicit în {intrebari} întrebări!")
            return intrebari
        elif r == "mic":
            dreapta = mijloc - 1
        else:
            stanga = mijloc + 1
    print("Hmm, răspunsurile tale se contrazic. Ai trișat! 😄")
    return None

joaca()
````

### 🎲 Simulări

````python
# Referință – Simulări
import random

def estimeaza_pi(n):
    in_cerc = 0
    for i in range(n):
        x = random.random()
        y = random.random()
        if x * x + y * y <= 1:
            in_cerc += 1
    return 4 * in_cerc / n

def probabilitate_sapte(n):
    sapte = 0
    for i in range(n):
        if random.randint(1, 6) + random.randint(1, 6) == 7:
            sapte += 1
    return sapte / n

PI = 3.141592653589793
for n in (100, 10000, 100000):
    p = estimeaza_pi(n)
    q = probabilitate_sapte(n)
    print(f"n={n:>6}  π≈{p:.4f} (eroare {abs(p - PI):.4f})   P(7)≈{q:.4f} (eroare {abs(q - 1/6):.4f})")
# Eroarea scade cam ca 1/√n: de 100 de ori mai multe aruncări → de ~10 ori mai precis.
````

### 🪵 Jocul NIM

````python
# Referință – Jocul NIM
def mutare_calculator(betisoare):
    luate = (betisoare - 1) % 4
    if luate == 0:
        luate = 1
    return luate

def citeste_mutare(betisoare):
    while True:
        text = input("Câte iei (1-3)? ").strip()
        if text.isdigit() and 1 <= int(text) <= 3 and int(text) <= betisoare:
            return int(text)
        print("Mutare invalidă.")

betisoare = 21
la_rand = "om"
while betisoare > 0:
    print(f"\nAu rămas {betisoare}: " + "|" * betisoare)
    if la_rand == "om":
        luate = citeste_mutare(betisoare)
    else:
        luate = mutare_calculator(betisoare)
        print("Calculatorul ia", luate)
    betisoare -= luate
    if betisoare == 0:
        if la_rand == "om":
            print("Ai luat ultimul bețișor: ai pierdut! 🤖")
        else:
            print("Calculatorul a luat ultimul: ai câștigat! 🎉")
    if la_rand == "om":
        la_rand = "calculator"
    else:
        la_rand = "om"
````

### 🧮 Calculatorul 2.0

````python
# Referință – Calculatorul 2.0
def aduna(a, b): return a + b
def scade(a, b): return a - b
def inmulteste(a, b): return a * b
def putere(a, b): return a ** b
def imparte(a, b):
    if b == 0:
        return None
    return a / b

def calculeaza(a, op, b):
    if op == "+": return aduna(a, b)
    if op == "-": return scade(a, b)
    if op == "*": return inmulteste(a, b)
    if op == "/": return imparte(a, b)
    if op == "**": return putere(a, b)
    return None

def citeste_numar(mesaj):
    while True:
        t = input(mesaj).strip()
        if t.replace(".", "", 1).replace("-", "", 1).isdigit():
            return float(t)
        print("Număr invalid.")

nr_operatii = 0
cel_mai_mare = None
while True:
    op = input("Operație (+ - * / ** sau q): ").strip()
    if op == "q":
        break
    if op != "+" and op != "-" and op != "*" and op != "/" and op != "**":
        print("Operație necunoscută.")
        continue
    a = citeste_numar("a = ")
    b = citeste_numar("b = ")
    r = calculeaza(a, op, b)
    if r is None:
        print("Nu se poate împărți la zero.")
        continue
    print(f"Rezultat: {r:g}")
    nr_operatii += 1
    if cel_mai_mare is None or r > cel_mai_mare:
        cel_mai_mare = r
print(f"Operații: {nr_operatii}. Cel mai mare rezultat: {cel_mai_mare}")
````

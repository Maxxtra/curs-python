### Tipuri de date
Fiecare valoare are un **tip**. Cele patru de care avem nevoie azi:

| Tip | Ce este | Exemple |
|---|---|---|
| `int` | număr întreg | `15`, `-3`, `2026` |
| `float` | număr cu virgulă (se scrie cu punct!) | `1.68`, `3.14`, `-0.5` |
| `str` | text (string), între ghilimele | `"Ana"`, `'salut'`, `"15"` |
| `bool` | adevărat sau fals | `True`, `False` |

Funcția `type()` îți spune tipul unei valori.

````python
print(type(nume))
print(type(varsta))
print(type(inaltime))
print(type(este_elev))
print(type("15"))   # atenție: cu ghilimele este text, nu număr!
````

### Reguli pentru numele variabilelor

- literele mari și mici contează: `Nume` și `nume` sunt variabile diferite

## 4. Operatori ⏱ 10:15–10:40

### Aritmetici
| Operator | Ce face | Exemplu | Rezultat |
|---|---|---|---|
| `+` `-` `*` | adunare, scădere, înmulțire | `7 * 3` | `21` |
| `/` | împărțire (rezultatul e mereu `float`) | `7 / 2` | `3.5` |
| `//` | câtul împărțirii (partea întreagă) | `7 // 2` | `3` |
| `%` | restul împărțirii | `7 % 2` | `1` |
| `**` | ridicare la putere | `2 ** 10` | `1024` |

````python
print(7 + 3, 7 - 3, 7 * 3)
print(7 / 2)     # 3.5
print(7 // 2)    # 3  (de câte ori intră 2 în 7)
print(7 % 2)     # 1  (ce rămâne)
print(2 ** 10)   # 1024
````

`//` și `%` par ciudate, dar sunt foarte utile:
- `numar % 2 == 0` → numărul e **par**
- `135 // 60` = 2 ore și `135 % 60` = 15 minute
- `1234 % 10` = 4 → ultima cifră a unui număr

Ordinea operațiilor e ca la matematică: mai întâi `**`, apoi `* / // %`, apoi `+ -`. Parantezele schimbă ordinea.

````python
print(2 + 3 * 4)      # 14
print((2 + 3) * 4)    # 20
print(135 // 60, "ore și", 135 % 60, "minute")
````

### De comparație
Rezultatul unei comparații este mereu `True` sau `False`.

| `==` egal | `!=` diferit | `<` `>` | `<=` `>=` |
|---|---|---|---|

⚠️ `=` pune o valoare într-o variabilă; `==` întreabă dacă două valori sunt egale.

````python
print(5 > 3)          # True
print(5 == 5)         # True
print(5 != 5)         # False
print("abc" == "abc") # True
print(10 <= 9)        # False
````

### Logici: `and`, `or`, `not`
- `A and B` → adevărat doar dacă **ambele** sunt adevărate
- `A or B` → adevărat dacă **cel puțin una** e adevărată
- `not A` → inversează

````python
varsta = 15
print(varsta >= 14 and varsta <= 18)   # este între 14 și 18?  True
print(varsta < 10 or varsta > 60)      # False
print(not varsta > 10)                 # False
````

## 5. Decizii: `if`, `elif`, `else` ⏱ 10:40–11:00

Până acum programele noastre făceau același lucru de fiecare dată. Cu `if`, programul **alege** ce să facă în funcție de o condiție.

```python
if conditie:
    # ce se întâmplă dacă e adevărat
else:
    # ce se întâmplă altfel
```

Două reguli de sintaxă:
1. după condiție pui **două puncte** `:`
2. liniile din interiorul lui `if` sunt **indentate** (4 spații). Așa știe Python ce aparține de `if`.

````python
varsta = int(input("Câți ani ai? "))

if varsta >= 18:
    print("Ești major.")
else:
    print("Ești minor.")

print("Această linie se afișează oricum, nu e indentată.")
````

Când ai mai multe cazuri, folosești `elif` (prescurtare de la *else if*). Python verifică de sus în jos și se oprește la **prima** condiție adevărată.

````python
ora = int(input("Cât e ceasul (0-23)? "))

if ora < 12:
    print("Bună dimineața!")
elif ora < 18:
    print("Bună ziua!")
else:
    print("Bună seara!")
````

Condițiile pot fi combinate cu `and` / `or`, iar `else` este opțional.

````python
temperatura = 27
ploua = False

if temperatura > 25 and not ploua:
    print("Vreme perfectă de ieșit afară.")

if temperatura < 0 or ploua:
    print("Ia-ți o haină.")
````

> 🧑‍🏫 **Note pentru profesor**
>
> Cea mai frecventă eroare de aici încolo: **IndentationError** (linia de sub `if` nu e indentată sau e indentată inegal). Spune-le: în notebook, după ce scrii `:` și apeși Enter, indentarea se face singură; nu o șterge. Arată și `==` vs `=` în condiție (`if varsta = 18:` → SyntaxError).
> 
> Un bun exemplu la tablă: „dacă e pauză și am bani, cumpăr un covrig” → `if pauza and bani >= 3:`.
> 
> Dacă ai timp: ce se întâmplă dacă în exemplul cu ora scriem `ora = 25`? (Intră pe `else`. Validarea intrărilor e o idee de bonus pentru elevii rapizi.)

## 8. Mini-proiect: Calculatorul 🧮 ⏱ 14:00–15:30

Construiești un calculator care cere două numere și un operator, apoi afișează rezultatul. Lucrezi în pereche. Mergi pas cu pas și **testează după fiecare pas**.

**Pasul 1.** Cere primul număr, operatorul (`+`, `-`, `*`, `/`) și al doilea număr. Numerele pot avea zecimale (`float`).

**Pasul 2.** Cu `if` / `elif`, calculează rezultatul pentru fiecare operator și afișează-l frumos: `12.0 + 3.0 = 15.0`.

**Pasul 3.** Împărțirea la zero nu se poate. Dacă operatorul e `/` și al doilea număr e `0`, afișează un mesaj în loc să lași programul să crape.

**Pasul 4.** Dacă utilizatorul scrie un operator pe care nu îl știi, afișează `Operator necunoscut`.

**Extra (alege ce vrei):**
- ⭐ adaugă `%`, `//` și `**`
- ⭐ rotunjește rezultatul la 2 zecimale: `round(rezultat, 2)`
- ⭐ adaugă operatorul `r` pentru radical: `numar ** 0.5`
- ⭐⭐ afișează un „meniu” la început, cu operatorii disponibili
- ⭐⭐ la împărțire, afișează și restul dacă numerele sunt întregi
- ⭐⭐⭐ transformă-l într-un calculator specializat: medie a notelor, conversie lei ↔ euro, calculul prețului cu reducere – ideea voastră

**La final (15:30)** fiecare pereche prezintă în 2 minute: ce face calculatorul, o demonstrație, ce a fost cel mai greu.

````python
# Soluție de referință – Pasul 1-4
a = float(input("Primul număr: "))
operator = input("Operatorul (+, -, *, /): ")
b = float(input("Al doilea număr: "))

if operator == "+":
    print(f"{a} + {b} = {a + b}")
elif operator == "-":
    print(f"{a} - {b} = {a - b}")
elif operator == "*":
    print(f"{a} * {b} = {a * b}")
elif operator == "/":
    if b == 0:
        print("Nu se poate împărți la zero!")
    else:
        print(f"{a} / {b} = {a / b}")
else:
    print("Operator necunoscut")
````

````python
# Soluție de referință – cu extra-uri (meniu, %, //, **, radical, rotunjire)
print("=== CALCULATOR ===")
print("Operatori: +  -  *  /  %  //  **  r (radical din primul număr)")

a = float(input("Primul număr: "))
operator = input("Operatorul: ")

if operator == "r":
    if a < 0:
        print("Nu pot calcula radical dintr-un număr negativ.")
    else:
        print(f"√{a} = {round(a ** 0.5, 2)}")
else:
    b = float(input("Al doilea număr: "))
    if operator == "+":
        rezultat = a + b
    elif operator == "-":
        rezultat = a - b
    elif operator == "*":
        rezultat = a * b
    elif operator == "/":
        if b == 0:
            print("Nu se poate împărți la zero!")
            rezultat = None
        else:
            rezultat = a / b
    elif operator == "%":
        rezultat = a % b
    elif operator == "//":
        rezultat = a // b
    elif operator == "**":
        rezultat = a ** b
    else:
        print("Operator necunoscut")
        rezultat = None

    if rezultat is not None:
        print(f"{a} {operator} {b} = {round(rezultat, 2)}")
````

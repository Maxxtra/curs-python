## 1. Bucla `while` – „repetă cât timp…” ⏱ 09:30–10:15

```python
while conditie:
    # se repetă cât timp condiția e adevărată
```

Ca la `if`: **două puncte** și **indentare**. Diferența: `if` rulează blocul o singură dată, `while` îl rulează din nou și din nou, cât timp condiția e `True`.

````python
i = 1
while i <= 5:
    print(i)
    i = i + 1
print("Gata!")
````

Orice buclă `while` are nevoie de **trei ingrediente**:

| Ingredient | În exemplu | Dacă lipsește… |
|---|---|---|
| start | `i = 1` | `NameError` |
| condiție | `i <= 5` | bucla nu se oprește corect |
| pas care schimbă ceva | `i = i + 1` | **buclă infinită** ♾️ |

> 🧑‍🏫 **Note pentru profesor**
>
> Fă la tablă un **tabel de urmărire** (trace table) pentru celula de mai sus, cu coloanele `i`, `i <= 5`, `afișează`. Completați-l împreună rând cu rând. Aceeași tehnică o folosesc apoi singuri la „urmărește execuția” și la debugging challenge.

````python
secunde = 10
while secunde > 0:
    print(secunde)
    secunde -= 1        # prescurtare pentru secunde = secunde - 1
print("Lansare! 🚀")
````

`while` e perfect când **nu știi dinainte** de câte ori trebuie să repeți. De exemplu, până când utilizatorul scrie ceva corect:

````python
nota = int(input("Nota (1-10): "))
while nota < 1 or nota > 10:
    print("Notă invalidă, mai încearcă.")
    nota = int(input("Nota (1-10): "))
print(f"Ai introdus nota {nota}.")
````

### `return` – rezultatul pe care ți-l dă înapoi funcția

### ⚠️ `print` nu este același lucru cu `return`
`print` doar **afișează** pe ecran. `return` **dă înapoi** o valoare pe care o poți pune într-o variabilă și folosi mai departe.

````python
def aria_print(l, L):
    print(l * L)

def aria_return(l, L):
    return l * L

x = aria_print(3, 4)     # afișează 12, dar x rămâne None (nimic)
y = aria_return(3, 4)    # nu afișează nimic, dar y primește 12
print("x =", x)
print("y =", y)
````

> 🧑‍🏫 **Note pentru profesor**
>
> Asta e confuzia numărul 1 la funcții. Regula simplă pentru ei: *dacă vrei să folosești rezultatul mai departe, ai nevoie de `return`*. Revine în debugging challenge (bug-ul B4).

````python
def este_par(n):
    return n % 2 == 0

def maxim(a, b):
    if a > b:
        return a
    return b            # ajungem aici doar dacă a > b a fost fals

print(este_par(10), este_par(7))
print(maxim(3, 9))
````

Funcțiile se combină ușor cu buclele:

````python
for n in range(1, 11):
    if este_par(n):
        print(n, "este par")
````

### E2. Suma cifrelor
Scrie funcția `suma_cifre(n)` care returnează suma cifrelor lui `n`. Exemplu: `suma_cifre(473)` → `14`.

Indiciu: ultima cifră este `n % 10`, iar `n // 10` „taie” ultima cifră. Repetă cât timp `n > 0`.

````python
# E2. Suma cifrelor
````

````python
# Soluție – E2
def suma_cifre(n):
    s = 0
    while n > 0:
        s += n % 10
        n = n // 10
    return s

print(suma_cifre(473))    # 14
print(suma_cifre(2026))   # 10
````

````python
# E3. Numără vocalele
````

### E4. Numere prime
Scrie funcția `este_prim(n)` care returnează `True` dacă `n` este prim (are exact doi divizori: 1 și el însuși). Apoi afișează toate numerele prime de la 2 la 50.

````python
# E4. Numere prime
````

````python
# Soluție – E4
def este_prim(n):
    if n < 2:
        return False
    for d in range(2, n):
        if n % d == 0:
            return False
    return True        # abia după ce am verificat toți divizorii

for n in range(2, 51):
    if este_prim(n):
        print(n, end=" ")
print()
# Pentru cei curioși: e suficient să verificăm d până la radical din n.
````

### E6. Palindrom
Scrie funcția `invers(text)` care returnează textul scris invers (`"Python"` → `"nohtyP"`), folosind un `for`. Apoi funcția `este_palindrom(text)`: un palindrom se citește la fel în ambele sensuri (`"cojoc"`, `"rotor"`).

Indiciu: pornești de la `rezultat = ""` și, pentru fiecare literă, faci `rezultat = litera + rezultat`.

````python
# E6. Palindrom
````

````python
# Soluție – E6
def invers(text):
    rezultat = ""
    for litera in text:
        rezultat = litera + rezultat
    return rezultat

def este_palindrom(text):
    return text == invers(text)

print(invers("Python"))
print(este_palindrom("cojoc"), este_palindrom("python"))
````

### E8 (bonus). Cel mai mare divizor comun
Scrie funcția `cmmdc(a, b)` cu algoritmul lui Euclid: cât timp `b` nu e 0, calculezi restul `r = a % b`, apoi `a` devine `b` și `b` devine `r`. La final răspunsul este `a`. Test: `cmmdc(48, 36)` → `12`.

````python
# E8 (bonus). Cel mai mare divizor comun
````

````python
# Soluție – E8 (bonus)
def cmmdc(a, b):
    while b != 0:
        r = a % b
        a = b
        b = r
    return a

print(cmmdc(48, 36))    # 12
print(cmmdc(17, 5))     # 1
````

### E9 (bonus). Conjectura lui Collatz
Pornești de la un număr `n`. Dacă e par, îl împarți la 2; dacă e impar, devine `3 * n + 1`. Repeți până ajungi la 1. Scrie funcția `pasi_collatz(n)` care returnează câți pași durează. Care număr de la 1 la 100 are nevoie de cei mai mulți pași?

````python
# E9 (bonus). Conjectura lui Collatz
````

````python
# Soluție – E9 (bonus)
def pasi_collatz(n):
    pasi = 0
    while n != 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        pasi += 1
    return pasi

cel_mai_bun = 1
for n in range(1, 101):
    if pasi_collatz(n) > pasi_collatz(cel_mai_bun):
        cel_mai_bun = n
print(f"{cel_mai_bun} are nevoie de {pasi_collatz(cel_mai_bun)} pași")   # 97, 118 pași
# Nimeni nu a demonstrat că orice număr ajunge la 1 – e o problemă deschisă din matematică.
````

### B9. Prim sau nu?
`este_prim(9)` ar trebui să dea `False`, iar `este_prim(7)` să dea `True`.

````python
def este_prim(n):
    for d in range(2, n):
        if n % d == 0:
            return False
        else:
            return True

print(este_prim(9))
print(este_prim(7))
````

**Bug:** `return True` e în `else`, deci funcția răspunde deja după primul divizor verificat (d = 2). `return True` trebuie mutat **după** buclă, neindentat față de `for`. Bonus: `este_prim(2)` dădea `None`.

````python
# Soluție – B9
def este_prim(n):
    if n < 2:
        return False
    for d in range(2, n):
        if n % d == 0:
            return False
    return True

print(este_prim(9))
print(este_prim(7))
````

### 🎯 Ghicește numărul PRO

````python
# Soluție – Ghicește numărul PRO
import random

MAX_INCERCARI = 7

def joaca_o_runda():
    secret = random.randint(1, 100)
    for incercare in range(1, MAX_INCERCARI + 1):
        ghici = int(input(f"Încercarea {incercare}/{MAX_INCERCARI}. Ghicește (1-100): "))
        if ghici == secret:
            print(f"🎉 Ai ghicit în {incercare} încercări!")
            return incercare
        elif ghici < secret:
            print("Prea mic")
        else:
            print("Prea mare")
    print(f"Ai pierdut. Numărul era {secret}.")
    return None

cel_mai_bun = None
raspuns = "d"
while raspuns == "d":
    rezultat = joaca_o_runda()
    if rezultat is not None and (cel_mai_bun is None or rezultat < cel_mai_bun):
        cel_mai_bun = rezultat
        print("Record nou!")
    raspuns = input("Mai joci? (d/n): ").lower()

print(f"La revedere! Cel mai bun scor: {cel_mai_bun}")
# De ce 7? Prin înjumătățire, 1–100 se rezolvă mereu în cel mult 7 încercări (2^7 = 128).
````

### 🎲 Simulator de zaruri

````python
# Soluție – Simulator de zaruri
import random

def arunca_zar():
    return random.randint(1, 6)

def simuleaza(n):
    duble = 0
    sapte = 0
    for i in range(n):
        z1 = arunca_zar()
        z2 = arunca_zar()
        if z1 == z2:
            duble += 1
        if z1 + z2 == 7:
            sapte += 1
    return duble, sapte

n = int(input("De câte ori arunc cele două zaruri? "))
duble, sapte = simuleaza(n)
print(f"Duble: {duble} ({duble * 100 / n:.1f}%)")
print(f"Suma 7: {sapte} ({sapte * 100 / n:.1f}%)")
print("Teoretic, ambele sunt 6 din 36 = 16.7%")
# Discuție: cu n = 10 procentele sar mult, cu n = 100000 se apropie de 16.7%.
# Legea numerelor mari – aceeași idee stă la baza antrenării modelelor de IA pe multe date.
````

### 🧮 Calculatorul 2.0

````python
# Soluție – Calculatorul 2.0
def aduna(a, b):
    return a + b

def scade(a, b):
    return a - b

def inmulteste(a, b):
    return a * b

def imparte(a, b):
    if b == 0:
        return None
    return a / b

def calculeaza(a, operator, b):
    if operator == "+":
        return aduna(a, b)
    elif operator == "-":
        return scade(a, b)
    elif operator == "*":
        return inmulteste(a, b)
    elif operator == "/":
        return imparte(a, b)
    return None

print("=== CALCULATOR 2.0 ===  (scrie q la operator ca să ieși)")
nr_operatii = 0
while True:
    operator = input("Operator (+ - * /): ")
    if operator == "q":
        break
    if operator not in ("+", "-", "*", "/"):
        print("Operator necunoscut")
        continue
    a = float(input("Primul număr: "))
    b = float(input("Al doilea număr: "))
    rezultat = calculeaza(a, operator, b)
    if rezultat is None:
        print("Nu se poate împărți la zero!")
    else:
        print(f"{a} {operator} {b} = {round(rezultat, 2)}")
        nr_operatii += 1
print(f"La revedere! Ai făcut {nr_operatii} operații.")
````

### ✖️ Antrenor pentru tabla înmulțirii

````python
# Soluție – Antrenor pentru tabla înmulțirii
import random

def pune_intrebare(nr):
    a = random.randint(2, 10)
    b = random.randint(2, 10)
    raspuns = int(input(f"[{nr}/10] Cât face {a} x {b}? "))
    if raspuns == a * b:
        print("✅ Corect!")
        return True
    print(f"❌ Greșit, era {a * b}.")
    return False

def mesaj(scor):
    if scor == 10:
        return "Perfect! 🏆"
    elif scor >= 7:
        return "Foarte bine!"
    return "Mai exersăm – data viitoare iese mai bine."

scor = 0
for i in range(1, 11):
    if pune_intrebare(i):
        scor += 1

print(f"Scor final: {scor}/10 – {mesaj(scor)}")
````

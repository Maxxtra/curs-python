## 1. Liste ⏱ 09:15–10:00

O **listă** ține mai multe valori, în ordine, într-o singură variabilă. Se scrie între paranteze drepte, cu virgulă între elemente.

```python
note = [9, 7, 10, 8, 6]
print(note)
print("Câte note:", len(note))
```

### Poziții (indici)
Fiecare element are o **poziție**, numărată **de la 0**.

| Poziție | 0 | 1 | 2 | 3 | 4 |
|---|---|---|---|---|---|
| `note` | 9 | 7 | 10 | 8 | 6 |
| Poziție de la final | -5 | -4 | -3 | -2 | **-1** |

```python
print(note[0])     # primul element
print(note[2])
print(note[-1])    # ultimul element

note[1] = 8        # schimbăm nota de pe poziția 1
print(note)
```

⚠️ `note[5]` dă **IndexError**: lista are 5 elemente, deci pozițiile merg de la 0 la 4.

### Adăugăm, verificăm, calculăm

```python
note.append(10)            # adaugă la final
print(note)

print(10 in note)          # există nota 10?  True / False
print("Suma:", sum(note))
print("Maxim:", max(note), "Minim:", min(note))
print("Media:", sum(note) / len(note))
print("Sortate:", sorted(note))
```

## 3. Dicționare ⏱ 10:15–10:45

O listă ține valori **în ordine**, găsite după poziție. Un **dicționar** ține perechi **cheie → valoare**, găsite după **nume**. Exact ca un dicționar adevărat: cauți cuvântul (cheia), găsești explicația (valoarea).

```python
elev = {"nume": "Ana", "varsta": 15, "clasa": "IX A"}

print(elev["nume"])
print(elev["varsta"])

elev["varsta"] = 16          # schimbăm o valoare
elev["oras"] = "București"   # adăugăm o cheie nouă
print(elev)
```

| Listă | Dicționar |
|---|---|
| `[9, 7, 10]` | `{"Ana": 9, "Mihai": 7}` |
| găsești după **poziție**: `note[0]` | găsești după **cheie**: `note["Ana"]` |
| `append` adaugă la final | `d[cheie] = valoare` adaugă sau schimbă |

```python
traduceri = {"mar": "apple", "caine": "dog", "pisica": "cat", "carte": "book"}

cuvant = "caine"
if cuvant in traduceri:                  # există cheia?
    print(f"{cuvant} = {traduceri[cuvant]}")

for ro, en in traduceri.items():         # parcurgem perechile
    print(f"{ro} → {en}")
```

⚠️ `traduceri["casa"]` dă **KeyError**, pentru că cheia nu există. Verifică întâi cu `in`, sau folosește `traduceri.get("casa", "nu știu")`.

### Tiparul „numărăm cu un dicționar”
Foarte folosit: de câte ori apare fiecare lucru.

```python
text = "abracadabra"
frecventa = {}
for litera in text:
    frecventa[litera] = frecventa.get(litera, 0) + 1
print(frecventa)
```

## 4. Cum organizăm o aplicație ⏱ 10:45–11:15

O aplicație adevărată are de obicei **trei părți**:

| Partea | Ce conține | Exemplu (catalog) |
|---|---|---|
| 🗂️ **Datele** | liste și dicționare | `catalog = [{"nume": ..., "nota": ...}]` |
| ⚙️ **Logica** | funcții care fac calcule, **fără `input`/`print`** | `media(catalog)`, `cel_mai_bun(catalog)` |
| 💬 **Interfața** | meniul, `input` și `print` | `while` cu „1. Adaugă, 2. Afișează…” |

Reguli simple:
1. **O funcție = o singură treabă**, cu un nume care spune ce face.
2. **Testezi fiecare funcție separat**, înainte să o pui în meniu.
3. Când lucrați în echipă, **fiecare ia o parte**: așa nu vă călcați pe picioare.

Mai jos e o aplicație mică, construită exact așa.

```python
# 🗂️ DATELE
catalog = []

# ⚙️ LOGICA
def adauga_elev(catalog, nume, nota):
    catalog.append({"nume": nume, "nota": nota})

def media(catalog):
    if len(catalog) == 0:
        return 0
    suma = 0
    for elev in catalog:
        suma += elev["nota"]
    return suma / len(catalog)

def cel_mai_bun(catalog):
    cel_mai_bun = catalog[0]
    for elev in catalog:
        if elev["nota"] > cel_mai_bun["nota"]:
            cel_mai_bun = elev
    return cel_mai_bun

# Testăm logica, înainte de meniu
adauga_elev(catalog, "Ana", 9)
adauga_elev(catalog, "Mihai", 7)
print(media(catalog))            # trebuie 8.0
print(cel_mai_bun(catalog))      # trebuie Ana
```

```python
# 💬 INTERFAȚA
def afiseaza_meniu():
    print()
    print("1. Adaugă elev")
    print("2. Afișează catalogul")
    print("3. Media și cel mai bun elev")
    print("0. Ieșire")

while True:
    afiseaza_meniu()
    optiune = input("Alege o opțiune: ")
    if optiune == "1":
        nume = input("Nume: ")
        nota = int(input("Nota: "))
        adauga_elev(catalog, nume, nota)
    elif optiune == "2":
        for elev in catalog:
            print(f"- {elev['nume']}: {elev['nota']}")
    elif optiune == "3":
        if catalog:
            print(f"Media: {media(catalog):.2f}")
            print(f"Cel mai bun: {cel_mai_bun(catalog)['nume']}")
        else:
            print("Catalogul e gol.")
    elif optiune == "0":
        print("La revedere!")
        break
    else:
        print("Opțiune necunoscută.")
```

> 🧑‍🏫 **Note pentru profesor**
>
> Arată că **meniul nu știe cum se calculează media**, doar cheamă funcția. Asta e ideea care face posibil lucrul în echipă: unul scrie `media`, altul scrie meniul, se întâlnesc la numele funcției. După-amiază vor împărți proiectul exact în aceste trei roluri.

### G2. Doar numerele pare
Scrie funcția `pare(lista)` care returnează o **listă nouă** doar cu numerele pare. `pare([1, 2, 3, 4, 10])` → `[2, 4, 10]`.

```python
# G2. Doar numerele pare
```

```python
# Soluție – G2
def pare(lista):
    rezultat = []
    for x in lista:
        if x % 2 == 0:
            rezultat.append(x)
    return rezultat

print(pare([1, 2, 3, 4, 10]))
print(pare([7, 9]))
```

### G3. Cel mai lung cuvânt

```python
# G3. Cel mai lung cuvânt
```

```python
# Soluție – G3
def cel_mai_lung(propozitie):
    cel_mai_lung = ""
    for cuvant in propozitie.split():
        if len(cuvant) > len(cel_mai_lung):
            cel_mai_lung = cuvant
    return cel_mai_lung

print(cel_mai_lung("Python este un limbaj extraordinar"))
```

### G4. Palindrom, varianta scurtă
Scrie `este_palindrom(text)` folosind `text[::-1]`. Ignoră literele mari și spațiile: `"Ele fac cafele"` trebuie să fie palindrom. Indiciu: `lower()` și `replace(" ", "")`.

```python
# G4. Palindrom, varianta scurtă
```

```python
# Soluție – G4
def este_palindrom(text):
    curat = text.lower().replace(" ", "")
    return curat == curat[::-1]

print(este_palindrom("Ele fac cafele"))
print(este_palindrom("Python"))
```

### G5. Frecvența cuvintelor
Scrie `numara_cuvinte(text)` care returnează un dicționar cu de câte ori apare fiecare cuvânt. Testează cu `"ana are mere ana are pere"`.

```python
# G5. Frecvența cuvintelor
```

```python
# Soluție – G5
def numara_cuvinte(text):
    frecventa = {}
    for cuvant in text.lower().split():
        frecventa[cuvant] = frecventa.get(cuvant, 0) + 1
    return frecventa

print(numara_cuvinte("ana are mere ana are pere"))
```

### G8 (bonus). Cifrul lui Cezar
Fiecare literă se înlocuiește cu cea aflată cu 3 poziții mai departe în alfabet (`a` → `d`, `x` → `a`). Scrie `cripteaza(text, pas)`. Indiciu: `alfabet = "abcdefghijklmnopqrstuvwxyz"`, `alfabet.index(litera)` și `%` 26.

```python
# G8 (bonus). Cifrul lui Cezar
```

```python
# Soluție – G8 (bonus)
alfabet = "abcdefghijklmnopqrstuvwxyz"

def cripteaza(text, pas):
    rezultat = ""
    for litera in text.lower():
        if litera in alfabet:
            pozitie = (alfabet.index(litera) + pas) % 26
            rezultat += alfabet[pozitie]
        else:
            rezultat += litera
    return rezultat

secret = cripteaza("salut python", 3)
print(secret)
print(cripteaza(secret, -3))      # decriptare: pasul invers
```

### ❓ Proiect A – Quiz

```python
# Referință – Quiz
import random

intrebari = [
    {"intrebare": "Câte zile are o săptămână?", "variante": ["5", "7", "10"], "corect": "7"},
    {"intrebare": "Care e cel mai mare ocean?", "variante": ["Atlantic", "Indian", "Pacific"], "corect": "Pacific"},
    {"intrebare": "Ce limbaj învățăm la acest modul?", "variante": ["Python", "Java", "C++"], "corect": "Python"},
    {"intrebare": "Cât face 7 * 8?", "variante": ["54", "56", "64"], "corect": "56"},
    {"intrebare": "Capitala Italiei?", "variante": ["Milano", "Roma", "Napoli"], "corect": "Roma"},
]

def normalizeaza(text):
    return text.strip().lower()

def verifica(intrebare, raspuns):
    return normalizeaza(raspuns) == normalizeaza(intrebare["corect"])

def mesaj_final(scor, total):
    procent = scor * 100 // total
    if procent == 100:
        return f"Perfect! {scor}/{total} 🏆"
    elif procent >= 60:
        return f"Foarte bine: {scor}/{total}"
    return f"{scor}/{total}. Mai încearcă!"

def o_runda():
    scor = 0
    random.shuffle(intrebari)
    for q in intrebari:
        print("\n" + q["intrebare"])
        for i, v in enumerate(q["variante"], start=1):
            print(f"  {i}) {v}")
        raspuns = input("Răspunsul tău (text sau număr): ")
        if raspuns.strip().isdigit() and 1 <= int(raspuns) <= len(q["variante"]):
            raspuns = q["variante"][int(raspuns) - 1]       # acceptă și numărul variantei
        if verifica(q, raspuns):
            print("✅ Corect!")
            scor += 1
        else:
            print(f"❌ Greșit, răspunsul era {q['corect']}.")
    print("\n" + mesaj_final(scor, len(intrebari)))
    return scor

record = 0
while True:
    scor = o_runda()
    record = max(record, scor)
    if input("Mai joci? (d/n): ").strip().lower() != "d":
        break
print(f"Record: {record}/{len(intrebari)}")
```

### 🪢 Proiect B – Spânzurătoarea

```python
# Referință – Spânzurătoarea
import random

cuvinte = ["python", "variabila", "functie", "lista", "dictionar", "algoritm",
           "calculator", "tastatura", "program", "internet"]
DESEN = ["😀", "🙂", "😐", "😕", "😟", "😨", "💀"]

def ascunde(cuvant, ghicite):
    return " ".join(litera if litera in ghicite else "_" for litera in cuvant)

def a_castigat(cuvant, ghicite):
    for litera in cuvant:
        if litera not in ghicite:
            return False
    return True

def citeste_litera(incercate):
    while True:
        litera = input("Literă: ").strip().lower()
        if len(litera) != 1 or not litera.isalpha():
            print("Scrie o singură literă.")
        elif litera in incercate:
            print("Ai mai încercat litera asta.")
        else:
            return litera

cuvant = random.choice(cuvinte)
ghicite = []
greseli = 0
MAX_GRESELI = 6

while greseli < MAX_GRESELI and not a_castigat(cuvant, ghicite):
    print(f"\n{DESEN[greseli]}  {ascunde(cuvant, ghicite)}   greșeli: {greseli}/{MAX_GRESELI}")
    litera = citeste_litera(ghicite)
    ghicite.append(litera)
    if litera not in cuvant:
        greseli += 1
        print("Nu e în cuvânt!")

if a_castigat(cuvant, ghicite):
    print(f"\n🎉 Ai câștigat! Cuvântul era {cuvant}.")
else:
    print(f"\n{DESEN[-1]} Ai pierdut. Cuvântul era {cuvant}.")
```

### 📊 Proiect C – Tracker de cheltuieli

```python
# Referință – Tracker de cheltuieli
cheltuieli = {"mancare": [], "transport": [], "distractie": []}

def adauga(cheltuieli, categorie, suma):
    if categorie not in cheltuieli:
        cheltuieli[categorie] = []
    cheltuieli[categorie].append(suma)

def total_categorie(cheltuieli, categorie):
    return sum(cheltuieli.get(categorie, []))

def total_general(cheltuieli):
    total = 0
    for categorie in cheltuieli:
        total += total_categorie(cheltuieli, categorie)
    return total

def categoria_maxima(cheltuieli):
    maxima = None
    for categorie in cheltuieli:
        if maxima is None or total_categorie(cheltuieli, categorie) > total_categorie(cheltuieli, maxima):
            maxima = categorie
    return maxima

def citeste_suma():
    while True:
        text = input("Suma (lei): ").replace(",", ".")
        try:
            suma = float(text)
            if suma > 0:
                return suma
        except ValueError:
            pass
        print("Scrie un număr pozitiv.")

BUGET = 200
while True:
    print("\n1. Adaugă cheltuială   2. Raport   0. Ieșire")
    optiune = input("Alege: ").strip()
    if optiune == "1":
        categorie = input("Categoria: ").strip().lower() or "altele"
        adauga(cheltuieli, categorie, citeste_suma())
    elif optiune == "2":
        total = total_general(cheltuieli)
        for categorie in cheltuieli:
            t = total_categorie(cheltuieli, categorie)
            procent = t * 100 / total if total else 0
            print(f"{categorie:12} {t:8.2f} lei  ({procent:.0f}%)")
        print(f"TOTAL: {total:.2f} lei · mai ai {BUGET - total:.2f} lei din buget")
        if total:
            print("Cele mai multe cheltuieli:", categoria_maxima(cheltuieli))
    elif optiune == "0":
        break
    else:
        print("Opțiune necunoscută.")
```

### 📒 Proiect D – Catalog

```python
# Referință – Catalog
catalog = {"Ana": [9, 10], "Mihai": [7, 8, 6]}

def adauga_nota(catalog, nume, nota):
    if nume not in catalog:
        catalog[nume] = []
    catalog[nume].append(nota)

def media(note):
    if len(note) == 0:
        return 0
    return sum(note) / len(note)

def clasament(catalog):
    perechi = []
    for nume, note in catalog.items():
        perechi.append((media(note), nume))
    return sorted(perechi, reverse=True)

def citeste_nota():
    while True:
        text = input("Nota (1-10): ").strip()
        if text.isdigit() and 1 <= int(text) <= 10:
            return int(text)
        print("Notă invalidă.")

while True:
    print("\n1. Adaugă notă   2. Mediile   3. Clasament   0. Ieșire")
    optiune = input("Alege: ").strip()
    if optiune == "1":
        nume = input("Elev: ").strip().title()
        if nume:
            adauga_nota(catalog, nume, citeste_nota())
    elif optiune == "2":
        for nume, note in catalog.items():
            print(f"{nume}: {note} → media {media(note):.2f}")
    elif optiune == "3":
        for loc, (m, nume) in enumerate(clasament(catalog), start=1):
            medalie = {1: "🥇", 2: "🥈", 3: "🥉"}.get(loc, "  ")
            print(f"{medalie} {loc}. {nume} – {m:.2f}")
    elif optiune == "0":
        break
    else:
        print("Opțiune necunoscută.")
```

### 🤖 Proiect E – Chatbot bazat pe reguli

```python
# Referință – Chatbot
import random

reguli = {
    "salut": ["Salut! 👋", "Bună! Ce mai faci?"],
    "ce faci": ["Învăț Python, ca și tine.", "Răspund la întrebări. Tu?"],
    "nume": ["Mă numesc PyBot."],
    "python": ["Python e limbajul meu preferat!", "Știai că Python a fost numit după un show de comedie?"],
    "scoala": ["Școala e mai faină cu programare. 😄"],
    "vreme": ["Nu am fereastră, dar sper că e soare."],
    "gluma": ["De ce s-a dus programatorul la doctor? Avea un bug. 🐛"],
    "multumesc": ["Cu plăcere!"],
}
NU_STIU = ["Nu am înțeles. Poți reformula?", "Hmm, despre asta nu știu încă."]
memorie = {"nume": None, "neintelese": 0}

def extrage_nume(mesaj):
    cuvinte = mesaj.split()
    for i in range(len(cuvinte) - 1):
        if cuvinte[i].lower() in ("numesc", "sunt"):
            return cuvinte[i + 1].strip(".,!").title()
    return None

def raspunde(mesaj):
    nume = extrage_nume(mesaj)
    if nume:
        memorie["nume"] = nume
        return f"Încântat, {nume}!"
    text = mesaj.lower()
    for cuvant_cheie in reguli:
        if cuvant_cheie in text:
            raspuns = random.choice(reguli[cuvant_cheie])
            if memorie["nume"]:
                raspuns = f"{memorie['nume']}, {raspuns[0].lower()}{raspuns[1:]}"
            return raspuns
    memorie["neintelese"] += 1
    return random.choice(NU_STIU)

print("PyBot: Salut! Scrie „pa” ca să ieși.")
while True:
    mesaj = input("Tu: ")
    if mesaj.strip().lower() == "pa":
        print(f"PyBot: La revedere! (n-am înțeles {memorie['neintelese']} mesaje 😅)")
        break
    print("PyBot:", raspunde(mesaj))
```

## 1. Bucla `while` – „repetă cât timp…” ⏱ 09:30–10:15

```python
while conditie:
    # se repetă cât timp condiția e adevărată
```

Ca la `if`: **două puncte** și **indentare**. Diferența: `if` rulează blocul o singură dată, `while` îl rulează din nou și din nou, cât timp condiția e `True`.

````python
i = 1
while i <= 5:
    print(i)
    i = i + 1
print("Gata!")
````

Orice buclă `while` are nevoie de **trei ingrediente**:

| Ingredient | În exemplu | Dacă lipsește… |
|---|---|---|
| start | `i = 1` | `NameError` |
| condiție | `i <= 5` | bucla nu se oprește corect |
| pas care schimbă ceva | `i = i + 1` | **buclă infinită** ♾️ |

````python
secunde = 10
while secunde > 0:
    print(secunde)
    secunde -= 1        # prescurtare pentru secunde = secunde - 1
print("Lansare! 🚀")
````

`while` e perfect când **nu știi dinainte** de câte ori trebuie să repeți. De exemplu, până când utilizatorul scrie ceva corect:

````python
nota = int(input("Nota (1-10): "))
while nota < 1 or nota > 10:
    print("Notă invalidă, mai încearcă.")
    nota = int(input("Nota (1-10): "))
print(f"Ai introdus nota {nota}.")
````

## 2. Bucla `for` și `range()` ⏱ 10:15–11:00

`for` repetă **pentru fiecare** element dintr-o secvență. Cea mai folosită secvență este `range()`:

| Scrii | Primești | Explicație |
|---|---|---|
| `range(5)` | 0, 1, 2, 3, 4 | de la 0, **fără** 5 |
| `range(1, 6)` | 1, 2, 3, 4, 5 | de la 1 până la 6, **fără** 6 |
| `range(0, 20, 5)` | 0, 5, 10, 15 | pas de 5 |
| `range(10, 0, -1)` | 10, 9, …, 1 | pas negativ = înapoi |

⚠️ Capătul din dreapta **nu este inclus**. E cea mai frecventă greșeală cu `range`.

````python
for i in range(1, 6):
    print(i)
````

````python
# list() doar ca să vedem toate numerele deodată (listele le facem data viitoare)
print(list(range(5)))
print(list(range(1, 6)))
print(list(range(0, 20, 5)))
print(list(range(10, 0, -1)))
````

### `for` sau `while`?

### `break` și `continue`

````python
for i in range(1, 11):
    if i == 7:
        break          # oprește bucla de tot
    print(i)
print("am ieșit din buclă")
````

````python
for i in range(1, 11):
    if i % 3 == 0:
        continue       # sare doar peste pasul ăsta și merge la următorul i
    print(i)
````

````python
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i} x {j} = {i * j}")
    print("---")
````

`def` doar **definește** funcția (o învață pe Python). Codul din ea rulează abia când o **apelezi**: `salut()`.

### Parametri – informația pe care i-o dai funcției

````python
def salut_pe(nume):
    print(f"Salut, {nume}!")

salut_pe("Ana")
salut_pe("Mihai")
````

### `return` – rezultatul pe care ți-l dă înapoi funcția

````python
def patrat(x):
    return x * x

a = patrat(4)
print(a)
print(patrat(3) + patrat(4))
````

### ⚠️ `print` nu este același lucru cu `return`
`print` doar **afișează** pe ecran. `return` **dă înapoi** o valoare pe care o poți pune într-o variabilă și folosi mai departe.

````python
def aria_print(l, L):
    print(l * L)

def aria_return(l, L):
    return l * L

x = aria_print(3, 4)     # afișează 12, dar x rămâne None (nimic)
y = aria_return(3, 4)    # nu afișează nimic, dar y primește 12
print("x =", x)
print("y =", y)
````

````python
def este_par(n):
    return n % 2 == 0

def maxim(a, b):
    if a > b:
        return a
    return b            # ajungem aici doar dacă a > b a fost fals

print(este_par(10), este_par(7))
print(maxim(3, 9))
````

Funcțiile se combină ușor cu buclele:

````python
for n in range(1, 11):
    if este_par(n):
        print(n, "este par")
````

### Parametri cu valoare implicită

````python
def putere(baza, exponent=2):
    return baza ** exponent

print(putere(5))        # exponent = 2, implicit
print(putere(2, 10))
````

### Descompunerea unei probleme în pași
O problemă mare se rupe în probleme mici, fiecare cu funcția ei. Exemplu: media a trei note și calificativul.

````python
def media(a, b, c):
    return (a + b + c) / 3

def calificativ(m):
    if m >= 9:
        return "Excelent"
    elif m >= 7:
        return "Bine"
    elif m >= 5:
        return "Suficient"
    return "Insuficient"

m = media(9, 8, 10)
print(f"Media {m:.2f}: {calificativ(m)}")   # :.2f = două zecimale
````

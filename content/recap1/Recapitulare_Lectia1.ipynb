## ⚙️ Pasul 0 – rulați celula de mai jos
Ea pregătește verificările automate. Nu trebuie să înțelegeți ce scrie în ea. Doar Shift + Enter.

````python
# ⚙️ Pregătirea verificărilor – rulați o dată, la început
_L = {}
_note_testate = set()
def _ok(m, c):
    _L[m] = chr(c)
    print(f"✅ Misiunea {m} e corectă! Ați primit litera „{chr(c)}”. Schimbați rolurile!")
def _nu(mesaj):
    print("❌ Încă nu. " + mesaj)

def verifica_1(nume1, nume2, varsta1, varsta2):
    if type(nume1) != str or type(nume2) != str or nume1 in ("", "...") or nume2 in ("", "..."):
        _nu("Numele trebuie să fie text, între ghilimele (și să fie numele voastre).")
    elif type(varsta1) != int or type(varsta2) != int:
        _nu("Vârstele trebuie să fie numere întregi, FĂRĂ ghilimele.")
    else:
        _ok(1, 80)

def verifica_2(tip_a, tip_b, tip_c, tip_d):
    corecte = [str(tip_a).strip().lower() == "int", str(tip_b).strip().lower() == "float",
               str(tip_c).strip().lower() == "str", str(tip_d).strip().lower() == "bool"]
    if all(corecte):
        _ok(2, 89)
    else:
        _nu(f"Aveți {sum(corecte)} din 4 corecte. Uitați-vă iar la tabelul cu tipuri.")

def verifica_3(bomboane, copii, fiecare, raman):
    if fiecare == bomboane // copii and raman == bomboane % copii:
        _ok(3, 84)
    elif fiecare == bomboane / copii:
        _nu("Ați folosit /, care dă un număr cu virgulă. Pentru bucăți întregi folosiți //.")
    else:
        _nu("Gândiți-vă: câte primește fiecare copil (//) și câte rămân (%)?")

def verifica_4(r1, r2, r3, r4, r5):
    corecte = [r1 == 8, r2 == 2.5, r3 == 3, r4 == 1, r5 is False]
    if all(corecte):
        _ok(4, 72)
    else:
        _nu(f"Aveți {sum(corecte)} din 5 corecte. Rulați celula cu print-uri și comparați.")

def verifica_5(numar, peste_doi_ani):
    if type(numar) != int:
        _nu("numar trebuie să fie număr întreg. Ce funcție transformă textul în număr întreg?")
    elif peste_doi_ani != numar + 2:
        _nu("peste_doi_ani trebuie să fie numar + 2.")
    else:
        _ok(5, 79)

def _calificativ_corect(nota):
    if nota >= 9: return "Excelent"
    if nota >= 7: return "Bine"
    if nota >= 5: return "Suficient"
    return "Insuficient"

def verifica_6(nota, calificativ):
    if calificativ != _calificativ_corect(nota):
        _nu(f"Pentru nota {nota} calificativul trebuia să fie altul. Verificați condițiile din if / elif.")
        return
    _note_testate.add(nota)
    if len(_note_testate) >= 2:
        _ok(6, 78)
    else:
        print(f"👍 Corect pentru nota {nota}! Acum schimbați nota cu una din altă categorie și rulați din nou ambele celule.")

def cuvant_secret():
    cuvant = "".join(_L.get(i, "_") for i in range(1, 7))
    print("Cuvântul vostru:", " ".join(cuvant))
    if "_" not in cuvant:
        print("🎉 GATA! Scrieți-vă numele pe tablă și arătați-i cuvântul profesorului.")
    else:
        print("Mai aveți misiuni de terminat. Locurile cu _ sunt misiunile lipsă.")

print("Verificările sunt gata. Puteți începe Misiunea 1!")
````

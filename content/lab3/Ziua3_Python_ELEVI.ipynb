## 3. Dicționare ⏱ 10:15–10:45

O listă ține valori **în ordine**, găsite după poziție. Un **dicționar** ține perechi **cheie → valoare**, găsite după **nume**. Exact ca un dicționar adevărat: cauți cuvântul (cheia), găsești explicația (valoarea).

```python
elev = {"nume": "Ana", "varsta": 15, "clasa": "IX A"}

print(elev["nume"])
print(elev["varsta"])

elev["varsta"] = 16          # schimbăm o valoare
elev["oras"] = "București"   # adăugăm o cheie nouă
print(elev)
```

| Listă | Dicționar |
|---|---|
| `[9, 7, 10]` | `{"Ana": 9, "Mihai": 7}` |
| găsești după **poziție**: `note[0]` | găsești după **cheie**: `note["Ana"]` |
| `append` adaugă la final | `d[cheie] = valoare` adaugă sau schimbă |

```python
traduceri = {"mar": "apple", "caine": "dog", "pisica": "cat", "carte": "book"}

cuvant = "caine"
if cuvant in traduceri:                  # există cheia?
    print(f"{cuvant} = {traduceri[cuvant]}")

for ro, en in traduceri.items():         # parcurgem perechile
    print(f"{ro} → {en}")
```

## 4. Cum organizăm o aplicație ⏱ 10:45–11:15

O aplicație adevărată are de obicei **trei părți**:

| Partea | Ce conține | Exemplu (catalog) |
|---|---|---|
| 🗂️ **Datele** | liste și dicționare | `catalog = [{"nume": ..., "nota": ...}]` |
| ⚙️ **Logica** | funcții care fac calcule, **fără `input`/`print`** | `media(catalog)`, `cel_mai_bun(catalog)` |
| 💬 **Interfața** | meniul, `input` și `print` | `while` cu „1. Adaugă, 2. Afișează…” |

Reguli simple:
1. **O funcție = o singură treabă**, cu un nume care spune ce face.
2. **Testezi fiecare funcție separat**, înainte să o pui în meniu.
3. Când lucrați în echipă, **fiecare ia o parte**: așa nu vă călcați pe picioare.

Mai jos e o aplicație mică, construită exact așa.

```python
# 🗂️ DATELE
catalog = []

# ⚙️ LOGICA
def adauga_elev(catalog, nume, nota):
    catalog.append({"nume": nume, "nota": nota})

def media(catalog):
    if len(catalog) == 0:
        return 0
    suma = 0
    for elev in catalog:
        suma += elev["nota"]
    return suma / len(catalog)

def cel_mai_bun(catalog):
    cel_mai_bun = catalog[0]
    for elev in catalog:
        if elev["nota"] > cel_mai_bun["nota"]:
            cel_mai_bun = elev
    return cel_mai_bun

# Testăm logica, înainte de meniu
adauga_elev(catalog, "Ana", 9)
adauga_elev(catalog, "Mihai", 7)
print(media(catalog))            # trebuie 8.0
print(cel_mai_bun(catalog))      # trebuie Ana
```

```python
# 💬 INTERFAȚA
def afiseaza_meniu():
    print()
    print("1. Adaugă elev")
    print("2. Afișează catalogul")
    print("3. Media și cel mai bun elev")
    print("0. Ieșire")

while True:
    afiseaza_meniu()
    optiune = input("Alege o opțiune: ")
    if optiune == "1":
        nume = input("Nume: ")
        nota = int(input("Nota: "))
        adauga_elev(catalog, nume, nota)
    elif optiune == "2":
        for elev in catalog:
            print(f"- {elev['nume']}: {elev['nota']}")
    elif optiune == "3":
        if catalog:
            print(f"Media: {media(catalog):.2f}")
            print(f"Cel mai bun: {cel_mai_bun(catalog)['nume']}")
        else:
            print("Catalogul e gol.")
    elif optiune == "0":
        print("La revedere!")
        break
    else:
        print("Opțiune necunoscută.")
```

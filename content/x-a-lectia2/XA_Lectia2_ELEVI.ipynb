## 1. Bucla `while`

`while` repetă cât timp condiția e adevărată. Orice `while` are **start**, **condiție** și **pas** care o apropie de fals. Când nu știi de câte ori se repetă, `while` e alegerea naturală.

````python
n = 4827
suma = 0
while n > 0:
    suma += n % 10      # ultima cifră
    n //= 10            # tăiem ultima cifră
print(suma)
````

### Algoritmul lui Euclid
Cel mai mare divizor comun: cât timp `b` nu e 0, `(a, b)` devine `(b, a % b)`. Răspunsul e `a`.

````python
a, b = 1071, 462
while b != 0:
    a, b = b, a % b
    print(a, b)
print("cmmdc =", a)
````

### Căutare prin înjumătățire
Calculatorul ghicește un număr ascuns între 1 și 1000 întrebând doar „mai mic sau mai mare?”. Câte întrebări îi trebuie în cel mai rău caz?

````python
secret = 737
stanga, dreapta = 1, 1000
pasi = 0
while stanga <= dreapta:
    mijloc = (stanga + dreapta) // 2
    pasi += 1
    if mijloc == secret:
        break
    elif mijloc < secret:
        stanga = mijloc + 1
    else:
        dreapta = mijloc - 1
print(f"Găsit {mijloc} în {pasi} pași")
````

### `break` și `else` pe buclă: căutăm ceva

````python
n = 91
for d in range(2, n):
    if n % d == 0:
        print(f"{n} nu e prim: se împarte la {d}")
        break
else:
    print(f"{n} e prim")     # rulează doar dacă bucla NU s-a oprit cu break
````

`else` după `for` pare ciudat, dar e exact „am căutat peste tot și n-am găsit”. Nu e obligatoriu: același lucru se face cu o variabilă `gasit = False`.

### Cât de repede?
Pentru `n` prim, bucla de mai sus face aproape `n` pași. Dar dacă `n = a × b`, unul dintre `a`, `b` e cel mult `√n`. Deci e suficient să verificăm divizorii până la `√n`:

````python
n = 1_000_003
pasi = 0
d = 2
prim = True
while d * d <= n:          # echivalent cu d <= radical din n, fără float
    pasi += 1
    if n % d == 0:
        prim = False
        break
    d += 1
if prim:
    print(n, "prim | pași:", pasi)
else:
    print(n, "compus | pași:", pasi)
````

## 3. Funcții

Funcțiile sunt cărămizile oricărui program mai mare: **o funcție = o singură treabă**, cu un nume clar, care **returnează** rezultatul (nu îl afișează).

````python
def suma_cifre(n):
    s = 0
    while n > 0:
        s += n % 10
        n //= 10
    return s

def este_prim(n):
    if n < 2:
        return False
    d = 2
    while d * d <= n:
        if n % d == 0:
            return False
        d += 1
    return True

# funcțiile se combină: numerele prime de 2 cifre cu suma cifrelor tot primă
for x in range(10, 100):
    if este_prim(x) and este_prim(suma_cifre(x)):
        print(x, end=" ")
print()
````

### B1. Prim sau nu?
Ar trebui să spună corect dacă un număr e prim.

````python
def este_prim(n):
    if n < 2:
        return False
    for d in range(2, n):
        if n % d == 0:
            return False
        else:
            return True

print(este_prim(7), este_prim(9), este_prim(2))
````

### B2. Suma cifrelor
`suma_cifre(4827)` ar trebui să dea `21`.

````python
def suma_cifre(n):
    s = 0
    while n > 0:
        s += n % 10
        n = n / 10
    return s

print(suma_cifre(4827))
````

### B3. Suma de la 1 la n

````python
def suma(n):
    s = 0
    for i in range(1, n):
        s += i
    return s

print(suma(100))
````

### B4. Factorialul
`factorial(5)` ar trebui să dea `120`.

````python
def factorial(n):
    produs = 0
    for i in range(1, n + 1):
        produs *= i
    return produs

print(factorial(5))
````
